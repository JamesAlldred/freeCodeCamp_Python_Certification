## Linear

```python
def linear_search(arr, target_value):
    for i in range(len(arr)):
        if arr[i] == target_value:
            return i
    return -1

print(f"Target value found at: {linear_search(numbers, 25)}")
```

## Merge Sort

```python
def merge_sort(arr):
    if len(arr) <= 1:
        return

    middle_index = len(arr) // 2
    left_section = arr[:middle_index]
    right_section = arr[middle_index:]

    left_section_index = 0
    right_section_index = 0
    sorted_index = 0

    merge_sort(left_section)
    merge_sort(right_section)

    while left_section_index < len(left_section) and right_section_index< len(right_section):
        if left_section[left_section_index] < right_section[right_section_index]:
            arr[sorted_index] = left_section[left_section_index]
            left_section_index += 1
        else:
            arr[sorted_index] = right_section[right_section_index]
            right_section_index += 1
        sorted_index += 1

    while left_section_index < len(left_section):
        arr[sorted_index] = left_section[left_section_index]
        left_section_index += 1
        sorted_index += 1

    while right_section_index < len(right_section):
        arr[sorted_index] = right_section[right_section_index]
        right_section_index += 1
        sorted_index += 1
    return

original = numbers.copy()   # keeps the unsorted order in case you need it again
merge_sort(numbers)         # sorts numbers directly, no assignment needed
assert numbers == sorted(original)   # silent if the sort is correct
print(f"Sorted numbers ready for binary search: {numbers}")
```

## Binary Search

```python
def binary_search(arr, target_value):
    path_to_target = []

    low = 0
    high = len(arr) - 1

    while low<= high:
        mid = (low + high) // 2
        middle_value = arr[mid]
        path_to_target.append(middle_value)

        if target_value == middle_value:
            return path_to_target, f"Value found at index: {mid}"

        elif target_value > middle_value:
            low = mid + 1
        else:
            high = mid - 1

    return [], "Value not found"

print(binary_search(numbers, 25))
```
