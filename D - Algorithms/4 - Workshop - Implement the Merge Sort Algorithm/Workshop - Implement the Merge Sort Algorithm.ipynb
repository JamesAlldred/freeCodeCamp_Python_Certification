# Workshop - Implement the Merge Sort Algorithm


Divide into halves until each subarray has one element, sort and merge.


```python
def merge_sort(array):
    if len(array) <= 1: # Base case: if the array has 1 or 0 elements, it is already sorted
        return

    #Find middle & split
    middle_point = len(array) // 2
    left_part = array[:middle_point]
    right_part = array[middle_point:]

    # Recursively sort both halves
    merge_sort(left_part)
    merge_sort(right_part)

    # Set indices for merging
    left_array_index = 0
    right_array_index = 0
    sorted_index = 0

    # Merge the sorted halves back into the original array
    while left_array_index < len(left_part) and right_array_index < len(right_part): # While there are elements in both halves
        if left_part[left_array_index] < right_part[right_array_index]: # Sort if smaller
            array[sorted_index] = left_part[left_array_index]
            left_array_index += 1 # Move to the next element in the left half
        else:
            array[sorted_index] = right_part[right_array_index]
            right_array_index += 1
        sorted_index += 1 # Move to the next position in the sorted array

    # While remaining elements in the left half, add them to the sorted array
    while left_array_index < len(left_part): 
        array[sorted_index] = left_part[left_array_index]
        left_array_index += 1
        sorted_index += 1
    
    # While remaining elements in the right half, add them to the sorted array
    while right_array_index < len(right_part):
        array[sorted_index] = right_part[right_array_index]
        right_array_index += 1
        sorted_index += 1

if __name__ == '__main__': # This block will only run if the script is executed directly, not if it is imported as a module
    numbers = [4, 10, 6, 14, 2, 1, 8, 5]
    print('Unsorted array: ')
    print(numbers)
    merge_sort(numbers)
    print('Sorted array: ')
    print(numbers)
```
