# Binary vs Linear Search


## Linear Search

- Starts @ beginning of list & iterates thru each item until target value


- If target value found: Index returned
- If isn't found: `-1` returned
  - (Because `-1` is not a valid index)


### Linear seach code:


```python
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
        return -1
```

### Straightforward but inefficient

- `O(n)` time complexity: Grows linearly with size of list
- `O`(1) space complexity: Constant space used


## Binary Search

- Divide list in half & check if target value is in middle of the list
  - Returns index
- If not, checks left or right half
- Continues to divide remaining parts into halves
  - Until TV found or returns `-1`


### Binary Search Code


```python
def binary_search(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid -1

        return -1
```

#### Code explanation:


- Start by identifying `low` and `high` index: Represents range of list to search
- Check `low` <= `high`
  - If `low` > `high`: Whole list searched & TV not found --> stop & return `-1`
  - If `low` <= `high`: Calc middle index `mid`
    - Check if TV @ `mid` index --> return `mid` if is
    - If not, check if value at `mid` is less thsn TV
      - If is, search right half of list: `low` = `mid` + 1
      - If not, search left half of list: `high` = `mid` - 1
- Repeat until TV found or `low` > `high` (TV not found)


### More efficient than linear

- `O(log n)` time complexity: Grows logarithmically with size of list (search space halved each iteration)
- `O(1)` space complexity: Constant space used
