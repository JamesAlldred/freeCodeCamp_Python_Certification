# Divide & Conquer and Merge Sort


- Recursively breaks down problems into small subproblems
  - Uses recursion & the merge sort algorithm


## Example Merge Sort Case

- We have this list: `[42, 37, 53, 17]`
- Goal: Sort list from smallest --> largest using merge sort algorithm
  <br></br> 1. Divide into 2 halves: `[42, 37]` | `[53, 17]` 2. Look at left half: `[42, 37]` - Divide into 2 halves: `[42]` | `[37]` - Divide in half again so each half has 1 element: `[42]` | `[37]` - Single item list sorted by default --> now need to merge these sublists into a sorted list: `[37, 42]` 3. Repeat for RHS & merge two sorted halves: `[17, 37, 42, 53]`


## Merge Sort Code


```python
def merge_sort(arr):
    if len(arr) <= 1: 
        return arr

    mid = len(arr) //2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    sorted_list = []
    i = 0
    j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            sorted_list.append(left[i])
            i += 1
        else:
            sorted_list.append(right[j])
            j += 1

    sorted_list.extend(left[i:])        
    sorted_list.extend(right[j:])

    return sorted_list
```

## Efficiency

- `O(n log n)` time complexity:
  - List continually divided in half (`O(log n)`)
  - Then merged (`O(n)`)
- `O(n)` space complexity: Not sorted in place, so new list created to hold sorted values
