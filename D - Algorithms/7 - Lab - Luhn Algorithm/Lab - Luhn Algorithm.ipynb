## Instructions


The Luhn algorithm, also known as the "modulus 10" or "mod 10" algorithm, is a simple checksum formula used to validate a variety of identification numbers, like credit card numbers. These are the steps to validate a number using the Luhn algorithm:

Starting from the right, and excluding the rightmost digit (the check digit), double the value of every other digit.
If the result of doubling a digit is greater than 9, sum the digits to get a single digit. Alternatively, you can subtract 9 from the result.
Take the sum of all the digits including the check digit.
If the sum of all the digits is a multiple of 10, then the number is valid; else it is not valid.
For example, consider the number 453914881. The steps to validate it using the Luhn algorithm are:

Account number 4 5 3 9 1 4 8 8 1
Double every other 4 10 3 18 1 8 8 16 1
Sum 2-char digits 4 1 3 9 1 8 8 7 1
Then sum all numbers, 4 + 1 + 3 + 9 + 1 + 8 + 8 + 7 + 1 = 42.
Since 42 is not a multiple of 10, the number is invalid.

## Code


```python
# Define initial function
def verify_card_number(numbers):

    # Remove whitespace and hyphens from the input string .replace(old,new)
    numbers = numbers.replace(" ", "")
    numbers = numbers.replace("-", "")
    print("Cleaned numbers:\n", numbers) # To check

    numbers_rev = numbers[::-1] # Slice: negative step reverses the string
    print("Reversed numbers:\n", numbers_rev) # To check
    
    total = 0 # Initialize total to 0. Created before the loops to not reset to 0 each time the loop runs

    # Loop 1: Double every second digit from the right, subtract 9 if the result is greater than 9, and add to total
    for n in range(1, len(numbers_rev), 2):
        digit = int(numbers_rev[n]) # [n]: access index (n) of value. int() for calcs
        operation_digit = 2 * digit
        
        if operation_digit > 9:
            operation_digit = operation_digit - 9
        print("Digit:", digit, "Operation result:", operation_digit)
        total += operation_digit

    # Loop 2: Add remaining digits to total (these aren't doubled)
    for n in range(0, len(numbers_rev), 2):
        total += int(numbers_rev[n])

    print("Total after Operations:\n", total) # To check

    # Check if total divisible by 10
    if total % 10 == 0: # %: modulus operator. Returns remainder. If total is divisible by 10, remainder is 0
        return "Status of number:\n" + "VALID!"
    else:
        return "Status of number:\n" + "INVALID!"
    
    

print(verify_card_number("453914889"), end ="\n\n")
print(verify_card_number("4111-1111-1111-1111"), end ="\n\n")
print(verify_card_number("1234 5678 9012 3456"), end ="\n\n")
```
