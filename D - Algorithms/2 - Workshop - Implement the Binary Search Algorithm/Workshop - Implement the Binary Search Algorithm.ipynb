# Implement the Binary Search Algorithm


Goal:
Search a list by spillting into halves then searching the half with the value


```python
def binary_search(search_list, value):
    path_to_target = [] # Track the path taken to the TV

    # Initialize low and high pointers. Spans the range of the entire list indices
    low = 0 
    high = len(search_list) - 1

    # Loop until the low pointer is greater than the high pointer (meaning the value is not found)
    while low <= high:
        mid = (low + high) // 2
        value_at_middle = search_list[mid] # Finds value at mid
        path_to_target.append(value_at_middle) # Add mid value to path track

        # Check if value found
        if value == value_at_middle:
            return path_to_target, f"Value found at index {mid}"

        # If not found, check which half to search & search
        elif value > value_at_middle:
            low = mid + 1 # Set the low index to the middle + 1 (search the right half)
        else:
            high = mid - 1 # Set the high index to the middle - 1 (search the left half)
    
    return [], "Value not found" # Return empty list and message if value not found

print(binary_search([1, 2, 3, 4, 5], 3))
print(binary_search([1, 2, 3, 4, 5, 9], 4))
print(binary_search([1, 3, 5, 9, 14, 22], 10))
```
