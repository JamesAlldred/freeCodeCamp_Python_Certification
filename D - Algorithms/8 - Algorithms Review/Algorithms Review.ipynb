# Algorithms Review


## Searching algorithms

Search for target within list of values


### 1 - Linear Search

- Iterates through list of items, checking each from beginning until found
  - If found: returns index
  - If not found: returns `-1` (invalid index)
- Inefficient for large lists
  - `O(n)` time complexity: Time grows linearly with size of list
  - `O(1)` space complexity: Doesn't need any more space to search through list


### 2 - Binary Search

- Divide list in half, checking if target in middle (returns index), left/right (repeat until middle or list empty)
- List must be sorted (asc or desc)
- More efficient --> ignores half w/o target
- If not found: returns `-1` (invalid index)
- `O(log n)` time complexity: Time grows logarithmically with size of list
- `O(1)` space complexity: Doesn't need any more space to search through list


### Linear vs Binary Search

- Binary better for large lists, but requires sorted list
- Linear time complexity: `O(n)` --> Time grows linearly with size of list
- Binary time complexity: `O(log n)` --> Time grows logarithmically with size of list


## Sorting algorithms & divide and conquer

Break down problem into smalller subproblems so easier to solve

- Recursion often used
- Divide and conquer used in many sorting algorithms (merge sort, quick sort)


### Merge Sort

Divide & conquer

- Recursively divide list into smaller sublist until each ahs one item (sorted)
- Merge sublists back together in order
- `O(n log n)` time complexity: Time grows logarithmically with size of list, and linear to merge sublists
- `O(n)` space complexity: Not in-place, needs extra space to merge sublists
