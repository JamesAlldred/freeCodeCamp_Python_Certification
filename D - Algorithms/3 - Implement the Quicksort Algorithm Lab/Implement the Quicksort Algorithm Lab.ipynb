```python
#Choose pivot value (first or last)
#Make 3 sublists:
    # Less than pivot
    # Equal
    # More
# Recursively call quick_sort to sort & concatenate into final list

def quick_sort(arr):
    if len(arr) <= 1: # Base Case: Needed to exit the recursion
        return arr
    
    pivot = arr[0] # First Value

    #Divide into 3 sections rel. to pivot
    less = quick_sort([value for value in arr if value < pivot]) # quicksort() on for loop that collects values less than pibot
    equal = [value for value in arr if value == pivot] # for loop collects values == pivot (already sorted as values ==)
    more = quick_sort([value for value in arr if value > pivot]) # # quicksort() on for loop that collects values less than pibot 
    
    
    return less + equal + more # Sorts values. Works within the three sections due to recursion

print(quick_sort([3, 6, 8, 10, 1, 2, 1,1,5,6,7,8,9,0,4,3,2,1,0])) 
```
