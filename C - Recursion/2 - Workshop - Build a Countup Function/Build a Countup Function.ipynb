# Workshop - Build a Countup function


```python
def countup(number):
    if number < 1: # Base case, returns empty list when number < 1
        return []
    count_list = countup(number - 1) # Recursive case, calls f() again with smaller argument: Moves towards base case
    count_list.append(number) # Appends number to list after recursive call returns
    return count_list

print(countup(5))
```
