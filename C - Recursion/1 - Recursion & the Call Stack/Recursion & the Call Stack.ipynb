# Recursion: Technique in which a function calls itself


- Each call works on smaller version of original problem until reaches condition to stop recursion (base case)
- Can solve problems with unknown depth
  - E.g. Traversing a tree structure
  - Or simpler: Counting down from given number


##### Recursive Countdown Example


```python
# recursive_countdown
def recursive_countdown(n):
    if n < 1: # Base case: Stops recursion when n < 1
        return
    print(n)

    return recursive_countdown(n-1) # Recursive case: Calls f() again with smaller argument 
recursive_countdown(5)
```

##### If `print()` after recursive call: f() prints numbers in ascending order


```python
# recursive_countdown printing - position of recursive call changes when numbers printed
def recursive_countdown(n):
    if n < 1: # Base case: Stops recursion when n < 1
        return
    
    recursive_countdown(n - 1) # Recursive case: Calls f() again with smaller argument 
    print(n)

recursive_countdown(5)
```

###### WHy does this happen? Call stack

- LIFO data structure that Python uses to track active f() calls - When f() called, Python adds frame for call to top of stack - Contains info like args & local variables - Removed from top of stack when f() returns
  <br></br>
- Every recursive f() needs base case & recursive case that moves towards it
  - When base case returns, earlier calls resume in LIFO orde
    - `print` statements after recursive calls are executed in reverse order of calls
