# Recursion Review

- Call f() repeatedly until base case is reached


##### Example code: Calculate factorial of a number


```python
def find_factorial(n):
    if n == 0: # Base case: Stops recursion when n == 0
        return 1
    return n * find_factorial(n - 1) # Recursive case
```

## Recursion info

- Handle something with unkown depth
- Call stack tracks function calls in recursive function
  - Each time called --> added to call stack
  - Base case --> function calls removed from call stack
- Carefully define base case (RecursionError)
- Useful for maths problems
