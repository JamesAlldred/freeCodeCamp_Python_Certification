# Tower of Hanoi Solver


```python
def hanoi_solver(n):
    # three rods as lists, the first holds every disk with the largest at the bottom
    rods = [list(range(n, 0, -1)), [], []]

    # turns the current rods into one line of text, e.g. "[3, 2] [] [1]"
    def snapshot():
        return " ".join(str(rod) for rod in rods)

    # the starting arrangement counts as the first line
    states = [snapshot()]

    # moves k disks from rod "source" to rod "target", using rod "spare" as parking
    def move(k, source, target, spare):
        # base case: nothing to move, so stop
        if k == 0:
            return

        # 1. park the k - 1 smaller disks on the spare rod
        #    (target and spare swap roles here)
        move(k - 1, source, spare, target)

        # 2. move the biggest of these k disks straight to the target
        disk = rods[source].pop()
        rods[target].append(disk)
        states.append(snapshot())  # record the arrangement after this move

        # 3. bring the k - 1 smaller disks from the spare rod onto the target
        #    (source and spare swap roles here)
        move(k - 1, spare, target, source)

    move(n, 0, 2, 1)  # rod indexes: 0 is the first rod, 2 is the last
    return "\n".join(states)


print(hanoi_solver(3))
    
```
