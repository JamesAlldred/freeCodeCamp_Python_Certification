# Classes & methods setup


## Define class & initialise instance attributes (`__init__`)

- Name of category
- Ledger: Empty list


```python
class Category:
    def __init__(self,name):
        self.name = name
        self.ledger = []
```

## Define first method - `deposit`

- Adds dictionary to empty ledger for `amount` and `description`
- `self`: links to instance
- `.append()`


```python
    def deposit(self, amount, description = ""):
        self.ledger.append({'amount': amount, 'description': description})
```

## Define second method - `withdraw`

- If `self.check_funds(amount)` is `True`: Adds negative amount & description to ledger list
- Otherwise `False`


```python
    def withdraw(self, amount, description = ""):
        if self.check_funds(amount):
            self.ledger.append({'amount': -amount, 'description': description})
            return True
        else:
            return False
```

## Define third method - `get_balance`

- Returns sum of `item` `amount`s in `self.ledger`


```python
    def get_balance(self):
        return sum(item["amount"] for item in self.ledger)
```

## Define fourth method - `transfer`

- Combines `withdraw` and `deposit`
- Transfers `amount` from the current category (`self`) to another specified `category`.


```python
    def transfer(self, amount, category):
        if self.check_funds(amount):
            self.withdraw(amount, f"Transfer to {category.name}")
            category.deposit(amount, f"Transfer from {self.name}")
            return True
        else:
            return False
```

## Define fifth method - `check_funds`

- Checks `amount` isn't more than balance using `get_balance()`
- Used to validate `withdraw` and `transfer`


```python
    def check_funds(self, amount):
        if amount > self.get_balance():
            return False
        else:
            return True
```

## Define Sixth method - `__str__`

- Dunder method - in built in background
- Returns formatted string representation of object
  - Category name
  - Ledger items
  - Total balance


```python
    def __str__(self):
        title = f"{self.name:*^30}\n"
        items = ""
        for item in self.ledger:
            desc = f"{item['description'][:23]:<23}"
            amt = f"{item['amount']:>7.2f}"
            items += f"{desc}{amt}\n"
        total = f"Total: {self.get_balance():.2f}"
        return title + items + total
```

# Chart


```python
def create_spend_chart(categories):
    # Calculate total withdrawals for each category across all ledgers.
    withdrawals = []
    for cat in categories:
        spent = sum(-item["amount"] for item in cat.ledger if item["amount"] < 0)
        withdrawals.append(spent)

    # Calculate percentage spent for each category, rounded to the nearest 10.
    total_spent = sum(withdrawals)
    if total_spent == 0:
        percentages = [0 for _ in categories]
    else:
        percentages = [int((w / total_spent) * 100) // 10 * 10 for w in withdrawals]

    # Build the header and vertical bars of the chart.
    chart = "Percentage spent by category\n"
    for i in range(100, -1, -10):
        chart += f"{i:>3}| "
        for pct in percentages:
            chart += "o  " if pct >= i else "   "
        chart += "\n"

    # Add the horizontal line separator.
    chart += "    " + "-" * (len(categories) * 3 + 1) + "\n"

    # Add category names as vertical labels below the chart.
    names = [cat.name for cat in categories]
    max_len = max(len(name) for name in names)

    for i in range(max_len):
        chart += "    "
        for name in names:
            char = name[i] if i < len(name) else " "
            chart += f" {char} "
        if i < max_len - 1:
            chart += " \n"
        else:
            chart += " "

    return chart
```
