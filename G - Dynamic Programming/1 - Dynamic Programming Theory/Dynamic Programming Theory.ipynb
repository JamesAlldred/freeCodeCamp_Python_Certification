# Dynamic Programming Theory
Algorithmic technique that solves complex problems by breaking down into subproblems
- Store results to avoid redundant calcs
- Normally take exponential time, reduces to polynomial time

## Core Principles
Works when two key conditions present

### 1 - Ovelapping Subproblems
- Same smaller problems appear multiple times when solving larger problems

### 2 - Optimal Substructure
- Optimal solutions has optimal solutions to subproblems

## Native recursion problem
- Climbing starecase with `n` steps and can climb 1 or 2 at a time.
- How many ways can you climb to the top?

- Has exp time complexity bc massive redundant calcs

### Dynamic programming solution
- 2 main approaches

#### 1 - Memoization (Top-Down)
Stores results of exxpensive func calls & returns cached result when same inputs occur again

```python
def climb_stairs_memo(n, memo={}):
    """Dynamic programming with memoization"""
    # Check if we've already calculated this value
    if n in memo:
        return memo[n]  # Return cached result - O(1) lookup!
    
    # Base cases
    if n <= 2:
        return n
    
    # Calculate once and store in memo for future use
    memo[n] = climb_stairs_memo(n-1, memo) + climb_stairs_memo(n-2, memo)
    return memo[n]
```

#### 2 - Tabulation (Bottom-Up)
Builds solution from gorund up, filling table with subproblem solutions
- Eliminates recursion completely

```python
def climb_stairs_tabulation(n):
    """Dynamic programming with tabulation"""
    if n <= 2:
        return n
    
    # Create array to store results for all steps from 0 to n
    dp = [0] * (n + 1)
    dp[1] = 1  # 1 way to reach step 1
    dp[2] = 2  # 2 ways to reach step 2
    
    # Build up the solution iteratively
    for i in range(3, n + 1):
        # Ways to reach step i = ways to reach (i-1) + ways to reach (i-2)
        dp[i] = dp[i-1] + dp[i-2]
    
    return dp[n]
```

## Real world applications

- GPS route optimisation
- Text processing
- Financial modelling
- Resource allocation

## When to use

- Problem can be broken down into overlapping subproblems
- Optimal structure
- Naive recursion would involve rep calcs
- Need to optimise for time complexity at cost of space complexity
