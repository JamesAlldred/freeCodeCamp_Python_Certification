```python
###Python Lists - mutable, zero based indexed. []

cities = ['Los Angeles', 'London', 'Tokyo']
print(cities[0]) #Access element, -1 for last element

#list() - creating list
print(list('abcdefg'))

#len()

#mutable - can update w/ index
update_list = list('12345678')
update_list[4] = 'INSERTED @ INDEX 4'
print(update_list)
#If index out of bounds - Index Error

#del - remove item from list
del update_list[7]
print(update_list)

#in - check if element in list
'2' in update_list #True

#Nested lists - access with index. Index twice to access indiv within nested
nested_list = [1, 2, 3, ['A', 'B', 'C']]
print(nested_list[3])
print(nested_list[3][2])

#Unpacking values from list - assign values from list to new variables
cricketer = ['Root', 66, 'Batter']
name, shirt_number, role = cricketer
print(name)
name, *details = cricketer #(*) collects any remaining elements
print(details)

#Slicing lists - access portion using (:). Start:Stop:Step. Stop exclusive
to_slice = ['1', '2', 'yellow', 'blue']
print(to_slice[0:3:2])
```

```python
###List methods

#.append() - add item to end of list
#.extend() - add multiple items to end of list
#.insert() - insert item at specific index
#.remove() - remove first occurrence of item in list
#.pop() - remove & return specific item from list. Defaults to last element
#.clear() - removes all items from list
#.sort() - sorts elements
#.sorted() - returns new sorted list instead of modifying original
#.reverse() - reverse order of elements in list
#.index() - finds first index where element can be found. If can't be found ValueError
```

```python
###Python Tuples - immutable (TypeError if try update). ()

#Access elements with [], use negative indexing to start at end. IndexError if too high index
#tuple() - create tuple
example_tuple = tuple('example')
print(example_tuple)

#in - check if item in tuple

#Unpacking tuple, (*) to collect remaining items
unpacking_tuple = ('Name', 'DOB', 'Height', 75)
name, DOB, *rest = unpacking_tuple
print(unpacking_tuple)

#slicing - use [start:stop]. stop exclusive.
#removing item TypeError - immutable

#When need to add/remove/alter - list
#Fixed & immutable data - tuple
```

```python
###Common Tuple Methods

#.count() - how many times item appears. False = 0. No arguments = TypeError

#.index() - what index item present. ValueError if not found
#Optional start index (inclusive) & optional end index (exclusive)
index_count_tuple = tuple('abcdefaa')
print(index_count_tuple)
index_count_tuple.count('a')
index_count_tuple.index('a', 1, 7)

#.sorted() - orders & returns new LIST. Modify with reverse, key: sorted(tuple_by_length, key = len)
```

Loops in Python - repeat a block of code for a set number of times

for loop - iterates over sequence & executes code for each item. Can nest
while loop - repeats code until condition False
break - exit loop immediately
continue - skips current iteration

for, while can have else - executed only when loop not termninated by break

```python
#for
handedness = ['Right arm', 'Left arm']
bowling_actions = ['Over' , 'Round', 'Runs']

for arm in handedness:
    for action in bowling_actions:
        print(arm, action)
```

```python
#while
dice = 5
guess = 0

while guess != dice:
    guess = int(input('Guess what the dice has rolled'))
    if guess != dice:
        print('Try again')
        
print('correct')
```

```python
#break
OS = ['Windows', 'Mac', 'Linux']

for system in OS:
    if system == 'Mac':
        break
    print(system)

#continue
OS = ['Windows', 'Mac', 'Linux']

for system in OS:
    if system == 'Mac':
        continue
    print(system)

#else
GhGs = ['greenhouse gases', 'CO2', 'WV', 'CH4', 'N20']

for gases in GhGs:
    if 'greenhouse gases' in GhGs:
        print(f'"{gases}" is a GhG')
    else:
        print(f'"{gases}" is not a GhG')
```

```python
###Ranges in loops

#range() - generate sequence of ints. range(start, stop, step). Stop mandatory, exclusive. Start default = 0
for num in range (0, 6,2):
    print(num)
#no arguments - TypeError
#negative step - decrementing order

#can also use range() with list()
numbers = list(range(2, 15, 3))
print(numbers)
```

```python
###enumerate() functions in Python

#enumerate() - iterate over sequence & keep track of index
#Takes iterable as argument & returns enumerate object (index + value)
#optional start argument, default 0

softwares = ['RStudio', 'Python', 'ArcGISPro']

for software in enumerate(softwares):
    print(software)

#can also be used outside for loop
to_enumerate = ['a','b']
print(list(enumerate(to_enumerate)))
```

```python
###zip() functions in Python

#zip() - iterate over multiple iterables in parallel

footballers = ['Haaland', 'Kane', 'Mbappe']
shirt_numbers = [67, 69, 21]

for ballers, number in zip(footballers, shirt_numbers):
    print(ballers)
    print(f'Shirt number: {number}')
```

```python
###List comprehensions in Python

#create new list in single line by combining loop & condition in square brackets
multiples_of_5 = [num for num in range (26) if num % 5 ==0]
print(multiples_of_5)
```

```python
###iterable methods

#filter() - filter elements from am iterable based on condition

to_filter = ['tree', 'bush', 'plain', 'mountain', 'river']

def short_word(to_filter):
    return len(to_filter) <= 4

short_word = list(filter(short_word, to_filter))
print(short_word)

#map() - apply function to each iterable & return new iterable with result

mph = [60, 65, 70, 75, 80, 85, 90]

def conv_kmph(speed):
    return(speed*8/5)

kmph = list(map(conv_kmph, mph))
print(kmph)

#sum() - get sum from iterable. Optional start argument
numbers = [1, 2, 3, 4]
print(sum(numbers, 15))
```

```python
###lambda functions -create function without name
#often used as argument to another function

numbers = [1, 2, 3, 4, 5]

numbers_squared = list(map(lambda x: x**2, numbers))
print(numbers_squared)
```
