```python
###Python Lists - mutable, zero based indexed. []

cities = ['Los Angeles', 'London', 'Tokyo']
print(cities[0]) #Access element, -1 for last element

#list() - creating list
print(list('abcdefg'))

#len()

#mutable - can update w/ index
update_list = list('12345678')
update_list[4] = 'INSERTED @ INDEX 4'
print(update_list)
#If index out of bounds - Index Error

#del - remove item from list
del update_list[7]
print(update_list)

#in - check if element in list
'2' in update_list #True

#Nested lists - access with index. Index twice to access indiv within nested
nested_list = [1, 2, 3, ['A', 'B', 'C']]
print(nested_list[3])
print(nested_list[3][2])

#Unpacking values from list - assign values from list to new variables
cricketer = ['Root', 66, 'Batter']
name, shirt_number, role = cricketer
print(name)
name, *details = cricketer #(*) collects any remaining elements
print(details)

#Slicing lists - access portion using (:). Start:Stop:Step. Stop exclusive
to_slice = ['1', '2', 'yellow', 'blue']
print(to_slice[0:3:2])
```

```python
###List methods

#.append() - add item to end of list
#.extend() - add multiple items to end of list
#.insert() - insert item at specific index
#.remove() - remove first occurrence of item in list
#.pop() - remove & return specific item from list. Defaults to last element
#.clear() - removes all items from list
#.sort() - sorts elements
#.sorted() - returns new sorted list instead of modifying original
#.reverse() - reverse order of elements in list
#.index() - finds first index where element can be found. If can't be found ValueError
```

```python
###Python Tuples - immutable (TypeError if try update). ()

#Access elements with [], use negative indexing to start at end. IndexError if too high index
#tuple() - create tuple
example_tuple = tuple('example')
print(example_tuple)

#in - check if item in tuple

#Unpacking tuple, (*) to collect remaining items
unpacking_tuple = ('Name', 'DOB', 'Height', 75)
name, DOB, *rest = unpacking_tuple
print(unpacking_tuple)

#slicing - use [start:stop]. stop exclusive.
#removing item TypeError - immutable

#When need to add/remove/alter - list
#Fixed & immutable data - tuple
```
