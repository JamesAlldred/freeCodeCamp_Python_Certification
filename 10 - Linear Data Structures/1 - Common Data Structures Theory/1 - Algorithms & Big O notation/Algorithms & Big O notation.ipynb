# Algorithms & big O notation


## Algorithm: Set of unambiguous instructions for solving a problem or performing a task.

- 2 key characteristics:

1. Can't continue indefintely. Finite numbe of steps.
2. Each step precise & unambiguous.

- Steps are independent of programming language
  - Need to be implemented in programming language to run
- Need to be efficient - time & space


## Big O notation: efficiency

- Describes worst case performance/growth rate of an algorithm
  - Growth rate: How resources needed inc with input size
- Not exact umber, but gives idea of how algorithm scales


### Input size & dominant terms

- `n`: input size
- Highest order term dominates growth rate as input size inc
  - Eg if `f(n) = 3n^2 + 5n + 2`, then `3n^2` will dominate as n inc


### Common time complexities

- Constant time: `O(1)`
  - Same run time regardless of input size
  - Accessing an element in an array
- Logarithmic time: `O(log n)`
  - Run time inc logarithmically as input size inc
  - Binary search
- Linear time: `O(n)`
  - Run time inc linearly as input size inc
  - for loop
- Quadratic time: `O(n^2)`
  - Run time inc quadratically as input size inc
  - Nestedn loops
- Log-linear time: `O(n log n)`
  - Run time inc log-linearly as input size inc
  - sorting algorithms
- Exponential `0(2^n)`, Factorial `O(n!)`
  - Inefficient in real world


### Space complexity

- How memory space required by an algorithm inc with input size
- Constant: `O(1)`
  - Same memory space regardless of input size
  - Eg uses a few variables to store values
- Linear: `O(n)`
  - Memory space inc linearly as input size inc
  - Eg creates and stores copy of list of length n
- Quadratic: `O(n^2)`
  - Memory space inc quadratically as input size inc
  - Eg creates and stores 2D array of size n x n
