# Maps, hash maps and sets


## Abstract Data Types (ADT)

- Conceptual representation of data type
  - Operations & properties
- Bluebrints that describe WHAT operations can be performed, not HOW
  - Separate interface from implementation


## Maps

- ADT that manages collections of key-value pairs & operations in specific & efficient way
  - Every value associated with unique key
  - Keys used to retrieve values
  - Keys must be unique, but values can be duplicated
- Defines important operations:
  - Insert key-value pair
  - Remove key-value pair
  - Retrieve value by key
  - Check if key exists
- Doesnt specify how, just lists as avaiable operations


## Hash Maps/hash tables

- Concrete implementation of map ADT
  - Use "hashing" to perform common operatione efficiently:
    - Generates a hash value (index) for each element using a hash function
    - Gen. based off KV pair and used to calc index in underlying array where KV pair stored
  - What if 2 keys generate same hash value? (Collision)
    - Chaining: each array index points to linked list, where all elements w/ same index stored
    - Open addressing: Search for next av index in array based oon predefined search seq
- Average time complexity for insert, delete, retrieve: `O(1)`
- Worst case: `O(n)` (Many collisions --> have to apply many collision resolution strategies)

- Average insertion space complexity: `O(1)`: Constant mem for each pair
  - Worst case: `O(n)`: Resizing of underlying array
- Removal generally `O(1)`


### Python dictionaries: Hash maps behind the scenes

```python
dictionary = {
    'A': 1, # Key: 'A', Value: 1
    'B': 2,
    'C': 3
}
# Or use dict() constructor
dictionary = dict(A=1, B=2, C=3)
```

Functions:

```python
dictionary['A'] # Retrieve value by key
dictionary['A'] = 10 # Update value by key
del dictionary['A'] # Remove key-value pair
'A' in dictionary # Check if key exists
dictionary.keys() # Get all keys
dictionary.values() # Get all values
dictionary.items() # Get all key-value pairs
```


## Sets

- Unordered collection of unique elements
- Guarantee no duplicates --> often used to remove duplicates from other data structures (list)
- Dynamic --> size can change as elements added/removed
  <br> </br>
- Average time complexity for adding, removing, getting length of set: `O(1)`
- Worst case: `O(n)` (Implemented as hash tables --> if many collisions --> have to apply many collision resolution strategies)
  <br> </br>
- Average space complexity for insertion: `O(1)` (Constant mem for each element)
- Worst case: `O(n)` (Resizing of underlying array)
- Removal generally `O(1)`


### Python sets

- Implemented as hash tables with only keys (no values)
  - Only immutable objects can be added to sets (hashable)
    - Hash values of mutable onjects can change --> can't be in sets as wouldnt be able to find them again

```python
numbers = {1, 2, 3, 4} # Can't do empty like this --> default to dict
# Or use set() constructor
numbers = set([1, 2, 3, 4]) # Use for empty set
```

<br> </br>
Functions:

```python
numbers.add(5) # Add element
numbers.remove(5) # Remove element, KeyError if not found
numbers.discard(5) # Remove element, no error if not found
numbers.pop() # Remove and return an arbitrary element from the set
numbers.clear() # Remove all elements from the set
5 in numbers # Check if element exists
```

<br> </br>


````python
# Set Operations
set_a = {1, 2, 3, 4}
set_b = {2, 3, 4, 5, 6}

print(set_a.union(set_b)) # All uniques
print(set_a.intersection(set_b)) # Common elements
print(set_a.symmetric_difference(set_b)) # Unique elements in either set
print(set_a.difference(set_b)) # Elements in set_a but not in set_b

# Or with equivalent operators
set_a | set_b
set_a & set_b
set_a ^ set_b
set_a - set_b
````

### Big O

- Average time complexity for adding, removing, testing membership: `O(1)`
- Worst case: `O(n)` (Hash tables collisions)
  - Could check is set subset/superset with `.issubset()` and `.issuperset()`
