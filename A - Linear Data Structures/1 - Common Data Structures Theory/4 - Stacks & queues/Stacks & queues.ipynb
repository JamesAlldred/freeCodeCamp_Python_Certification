# Stacks & queues

- Linear data structures that follow specific rules for adding & removing elements


## Stacks: LIFO (last in, first out) data structure

- Last element added is the first to be removed
- Two ends: top & bottom
  - Elements added & removed from top

- push: adding element
- pop: removing element
- Typically `O(1)` space & time complexity


## Queues: FIFO (first in, first out) data structure

- First element added is the first to be removed
- Two ends: front & back
  - Elements added to back, removed from front

- enqueue: adding element to back
- dequeue: removing element from front
  T
- Typically `O(1)` space & time complexity
