# 6 - Classes and Objects Review


## Definitions

- Class: Blueprint for creating objects
  - Reusable template
  - Placeholder for values when called
  - Make with `class ClassName:`
  - `Recep`

- Object: Instance of class
  - Actual data
  - Make with `obj_name = Classname('value')`
  - `receptionist`

- Attribute: Variable holding object-specific data
  - Each objct own attribute
  - `self.name`
    - `self`: represents specific instance
      - bind attributes to instance (`self.sender = sender`)
      - access instance methods (inside `display_full_email()` method `self.mark_as_read()` calls method on same object)
      - passes automatically (`self` passed behind scenes already in `tory.send_email()`)

- Method: Func def w/in class that operates on object
  - call with `obj_name.method_name()`

- Parameter: Variable in function/method
  - Placeholder for values when called
  - `self, name`
- Argument:Actual variable passed to function/method
  - `self`


### Examples


````python
class Recep: # Class
    role = 'admin' # Class attribute: defined directly in class & shared by all instances

    def __init__(self, name): # Dunder method with parameters. Initialise attributes
        self.name = name # Instance attribute: defined in __init__ & unique to each object/instance  

    def CSN(self): # Method
        return f'{self.name}, the receptionist,often says computer says no.'

receptionist = Recep("James") # Object

print(f"Receptionist is largely an {receptionist.role} role. {receptionist.CSN()}") # Method & attribute call
````

### Dunder (Magic) Methods

- Special methods for in-build python operations
- Start/end with `__`

- Called automatically in background w/ certain actions:
  - Arithmetic ops
  - String ops
  - Comparison ops
  - Iteration ops

- `__init__`: Initialises object attributes
  - Instance attributes: defined in `__init__` & unique to object/instance
  - Structure: `self.attribute = parameter`
    ```python
    class ClassName:
        def __init__(self, param_1, param_2, ..., param_n):
            self.attrib_1 = param_1
            self.attrib_2 = param_2
            self.attrib_n = param_n
    ```
