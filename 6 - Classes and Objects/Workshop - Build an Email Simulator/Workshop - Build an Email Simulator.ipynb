## `import datetime` module

- Allows timestamped emails
- Uses `datetime.datetime.now()` to give current date/time
- Uses `strftime('%Y-%m-%d %H:%M')` to format order & separators


```python
import datetime
```

## `class`, `objects`, `attributes`, `methods`, `parameters`, `arguments`

- Class (`Email`): Blueprint for creating objects.
  - Reusable templates
  - Takes parameters
  - Placeholder for values when called

- Object (`tory = User('Tory')`): Instance of class
  - Actual data
  - Make with `object_name = ClassName('value')`

- `__init__()`: initialises object attributes
  - format: `self.attribute = parameter`
  - Dunder (Magic) Method: Special methods for in-built Python operations
  - `self`: represents specific instance
    - bind attributes to instance (`self.sender = sender`)
    - access instance methods (inside `display_full_email()` method `self.mark_as_read()` calls method on same object)
    - passes automatically (`self` passed behind scenes already in `tory.send_email()`)

- Method (`display_full_email()`): Function def w/in class that operates on the object
  - call later with `object.method()`

- Parameter (`self, sender, receiver`...): Placeholder variable in func/method
- Argument (`Tory`): Actual variable passed to func/method
- Attribute (`self.sender`): Variable that holds object-specific data
  - Each object own `sender, receiver, subject...` attribute


```python
class Email: # Blueprint for creating email objects
    def __init__(self, sender, receiver, subject, body): # initialises attributes of objects
        self.sender = sender
        self.receiver = receiver
        self.subject = subject
        self.body = body
        self.timestamp = datetime.datetime.now()
        self.read = False

    def mark_as_read(self): # Method to mark the email as read
        self.read = True

    def display_full_email(self): # Method to format & print the email details
        self.mark_as_read()
        print('\n--- Email ---')
        print(f'From: {self.sender.name}')
        print(f'To: {self.receiver.name}')
        print(f'Subject: {self.subject}')
        print(f"Received: {self.timestamp.strftime('%Y-%m-%d %H:%M')}")
        print(f'Body: {self.body}')
        print('------------\n')

    def __str__(self): # Dunder method to return a string representation of the email object
        status = 'Read' if self.read else 'Unread'
        return f"[{status}] From: {self.sender.name} | Subject: {self.subject} | Time: {self.timestamp.strftime('%Y-%m-%d %H:%M')}"
```

```python
class User: # Blueprint for creating user objects
    def __init__(self, name): #  Initialises attributes of objects
        self.name = name
        self.inbox = Inbox() # (Calls Inbox class after User created later on)

    def send_email(self, receiver, subject, body): # Method to create email object + send to receiver inbox
        email = Email(sender=self, receiver=receiver, subject=subject, body=body)
        receiver.inbox.receive_email(email)
        print(f'Email sent from {self.name} to {receiver.name}!\n')

    def check_inbox(self): # Method to display emails in inbox
        print(f"\n{self.name}'s Inbox:")
        self.inbox.list_emails()

    def read_email(self, index): # Method to read specific email in inbox
        self.inbox.read_email(index)

    def delete_email(self, index): # Method to delete specific email in inbox
        self.inbox.delete_email(index)
```

```python
class Inbox: # Blueprint for creating inbox objects
    def __init__(self): 
        self.emails = [] # sets attribute to empty list to store email objects

    def receive_email(self, email): # Method to add email object to inbox
        self.emails.append(email)

    def list_emails(self): # Method to display a list of emails in the inbox
        if not self.emails:
            print('Your inbox is empty.\n')
            return
        print('\nYour Emails:')
        for i, email in enumerate(self.emails, start=1): # enumerate() function to get index + email object
            print(f'{i}. {email}')

    def read_email(self, index): # Method to read a specific email in the inbox
        if not self.emails:
            print('Inbox is empty.\n')
            return
        actual_index = index - 1 # Adjust for 0-based indexing
        if actual_index < 0 or actual_index >= len(self.emails):
            print('Invalid email number.\n')
            return
        self.emails[actual_index].display_full_email() # displays full email for index of self.emails list defined earlier

    def delete_email(self, index): # Method to delete a specific email in the inbox
        if not self.emails:
            print('Inbox is empty.\n')
            return
        actual_index = index - 1
        if actual_index < 0 or actual_index >= len(self.emails):
            print('Invalid email number.\n')
            return
        del self.emails[actual_index] # Deletes the email object at the specified index from the inbox
        print('Email deleted.\n')
```

```python
def main(): # Main function to demonstrate the email simulator
    tory = User('Tory') # Creates a user object named Tory
    ramy = User('Ramy') # Creates a user object named Ramy

    #Call the methods to send, check, read, and delete emails
    tory.send_email(ramy, 'Hello', 'Hi Ramy, just saying hello!') # 
    ramy.send_email(tory, 'Re: Hello', 'Hi Tory, hope you are fine.')
    ramy.check_inbox()
    ramy.read_email(1) # Reads the first email in Ramy's inbox - corrected for zero based indexing
    ramy.delete_email(1)
    ramy.check_inbox()
if __name__ == '__main__': # Ensures that the main function is called when the script is run directly
    main()
```
