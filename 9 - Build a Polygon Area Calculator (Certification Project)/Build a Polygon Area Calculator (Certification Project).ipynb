# Build a Polygon Area Calculator


## Explanation of when to use just (self) vs (self, height, width...):

- If method needs additional parameters (height, width, etc.): you include after 'self'.


## Parent class


```python
import math

class Rectangle:
    def __init__(self, width, height): # Three parameters: self, width, height
        self.width = width
        self.height = height
    
    def set_width(self, width): # Two parameters: self, width. Need to pass in new width value when calling the method  
        self.width = width
    
    def set_height(self, height): # Need height as for calling method, so two parameters: self, height
        self.height = height

    def get_area(self): # Just self because no new values, just attributes
        area = self.width * self.height
        return area

    def get_perimeter(self): 
        perimeter = 2 * (self.width + self.height)
        return perimeter
    
    def get_diagonal(self):
        diagonal = math.sqrt(self.width**2 + self.height**2)
        return diagonal
    
    def get_picture(self):
        if self.width > 50 or self.height > 50:
            return "Too big for picture."
        else:
            return ("*" * self.width + "\n") * self.height # "*" * self.width: String of '*' repeated width times. Whole expression repeated height times.

    def get_amount_inside(self, shape):
        return (self.width // shape.width) * (self.height // shape.height) # //: Integer division. Returns how many times shape can fit inside the rectangle.

    def __str__(self):
        return f"Rectangle(width={self.width}, height={self.height})"
```

## Subclass


```python

class Square(Rectangle):
    def __init__(self, side): # Two parameters: self, side. Need to pass in new side value when calling the method
        super().__init__(side, side) # Call the parent class's __init__ method with side as both width and height

    def set_width(self, width):
        self.set_side(width)
    
    def set_height(self, height):
        self.set_side(height)

    def set_side(self, side): # Two parameters: self, side. Need to pass in new side value when calling the method
        self.width = side
        self.height = side

    def __str__(self):
        return f"Square(side={self.width})"        
```

## Using the code


### On rectangle (parent)


```python
# Making rectangle object and changing dimensions
rectangle = Rectangle(10, 5)
rectangle.set_width(7)
rectangle.set_height(3)

# Getting area, perimeter, diagonal, and picture of rectangle
rectangle_area = rectangle.get_area()
print(f"Rectangle Area: {rectangle_area}")

rectangle_perimeter = rectangle.get_perimeter()
print(f"Rectangle Perimeter: {rectangle_perimeter}")

rectangle_diagonal = rectangle.get_diagonal()
print(f"Rectangle Diagonal: {rectangle_diagonal}")

rectangle_picture = rectangle.get_picture()
print(rectangle_picture)

# Making new rectangle to use get_amount_inside method
small_rectangle = Rectangle(2, 1)
amount_inside = rectangle.get_amount_inside(small_rectangle)
print(f"Amount of small rectangles inside the larger rectangle: {amount_inside}")

# Using __str__
print(rectangle)
```

### On Square (sub)


```python
# Making square object
square = Square(10)
square.set_side(3)
print(square)

#How many squares can fit inside the rectangle
sqares_inide = rectangle.get_amount_inside(square)
print(f"Amount of squares inside the rectangle: {sqares_inide}")
```
