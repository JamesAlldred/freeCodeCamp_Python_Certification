# Priority Queues and Heaps


## Priority Queue


Abstract Data Type (ADT)

Normal queues, stacks are FIFO, LIFO --> only consider order of insertion

Priority queues take element priority into account

- Determines removal order
- Usually highest priority first
- V helpful for shortest path btwn locations, scheduling tasks, simulating traffic, compressing data, managing networks
- Implemented using heap data structure


## Heap


Tree data structure with heap property

- Determines rel btwn parent and child nodes
- Two primary types


### 1 - Max heap

Value of node >= children value


### 2 - Min heap

Value of node <= children value


### Implementing Heaps as Arrays

Access parent and child nodes efficiently

- Simplifies logic for accessing
  - If maintains binary tree structure array implementation only needs simple maths operations bases on indices to locate elements in memory


#### Using `hepq` module


Operates directly on Python lists following specific steps that work w/ elements as if list was a heap --> rpeserves heap property


```python
import heapq
my_heap = []
heapq.heappush(my_heap, 9) # Add element to heap
heapq.heappop(my_heap) # Get element with highest priority
heapq.heappushpop(my_heap, 15) # Combine push and pop into one operation

heapq.heapify(my_heap) # Convert list into heap
```

##### Sorting by Priority
Store tuples as `(priority, element)`
- As tuples compared elem to eleme L --> R, priorities compared first & decided
- As min heap: lower values higher priority

If need elem with same priotiy to be returned in order of insertion, can use counter as second element in tuple `(priority, counter, element)`

```python
my_heap = []

heapq.heappush(my_heap, (3, "A"))
heapq.heappush(my_heap, (2, "B"))
heapq.heappush(my_heap, (1, "C"))
```

### Time and space complexity

Avg & worst case time complexities for insert & extract min/max value `O(log n)` bc number of swaps needed prop to height of leap (`log(n)`)

Peek: Average/Worst `O(1)` (retrieves min/max without removal)

Heapify: Average/Worst `O(n)` (builds heap from unsorted list)

Search / Delete (Arbitrary): Average/Worst `O(n)` (may traverse entire heap)

Space Complexity: `O(n)`  (stores n elements plus minor array/list overhead)

Core Application: Priority queues (efficient access to highest/lowest priority elements, e.g., Dijkstra's shortest path algorithm)
