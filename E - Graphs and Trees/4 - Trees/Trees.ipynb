# Tree
Specific type of graph
- No loops/cycles (paths where start and end nodes are the same)
- Connected (there is a path between any two nodes)

Non-linear data structures that order nodes in hierarchy
- Nodes can have chiodren, sibings, parents
- Root node --> top of tree, no parent
    - Will start traversing data structure here (BFS or DFS)




## Tree Terminology

- Parent node --> directly connected to nodes below
- Child node --> directly connected to nodes above
- Leaf node --> no children

- Depth --> length of path from root to node
- Height --> length of path from node to leaf
- Degree --> number of children a node has

## Tree Types
Many including Binary, Bineary Search, AVL, Red-Black, B-Trees

### Binary & Binary Search Trees
Two most common tree types

Binary tree --> each node can have at most 2 children (left + right)

Binary search tree --> binary tree with particular ordering
- Subtree: section of tree that is a tree itself
- Ordering property of BST establishes for every node:
    - All nodes in left subtree have values less than parent node
    - All nodes in right subtree have values greater than parent node
- Left and right subtrees must also be binary search trees
- This makes search, insertion, deletion efficiient if tree is balanced
    - Balanced tree --> height of left and right subtrees v similar to make sure operations efficient

##### Tries

Tree data structures used to store a set of rings
- Also known as prefix trees --> bc very effiecient for finding strings based on prefixes
- Each node in trie rep single character of string
    - Root doesnt represent any charact, think as empty string
    - As traverse trie down from root, path to node defines specific prefix
        - To find word --> follow prefix until reach node with word
        - Nodes that rep complete words marked as "end of word" nodes
- Worst time complexity for searching, inserting, deleting is `O(L)` where L is length of string
- Good at: many strings with same prefix (paths overlap, prefix stored once)
    - Autocomplete, spell check
- Bad at: strings with many unique characters (many nodes with only 1 child, long paths)
