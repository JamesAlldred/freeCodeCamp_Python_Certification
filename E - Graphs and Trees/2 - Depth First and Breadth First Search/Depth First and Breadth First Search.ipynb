# Depth First and Breadth First Search

Traversing the data structure - visiting each node


Systematically visit each node to make sure no skips
BFS & DFS algorithms traverse graphs & find paths between nodes

- Define order of node visits


## BFS

Visits all neighbouring nodes before moving to the next level in graph


Shortest path between nodes in unweighted graph

- Analyses all nodes at each level --> find path with fewest edges first
  Commonly implemented using queue data structure to track visited nodes (FIFO)


### How works:

1. Start at specific node
2. Mark as visited --> added to queue
3. While queue not empty:
   - Current node removed from queue (dequeued)
   - For all neighbours:
     - If not visited, mark as visited and added to queue

Because needs to store queue in mem --> large space requirements

- Esp for graphs with many nodes on same level


### Example in tree graph




#### Step 1:

Start at root (A)
Add A to queue and mark as visited

- Queue: [A]
- Visited: {A}


#### Step 2:

Dequeue A

Add unvisited children to queue and mark as visited

- Queue: [B, C]
- Visited: {A, B, C}

Order for nodes in same level addes to queue def by data structure implementation & order of edges stored in graph

- If implemmentation consistent --> order node traversal w/in level won't affect algorithm correctness
- Still visit every node level by level


#### Step 3:

Dequeue B

Add unvisited children to queue and mark as visited

- Queue: [C, D, E]
- Visited: {A, B, C, D, E}


#### Step 4:

Dequeue C

Add unvisited children to queue and mark as visited

- Queue: [D, E, F, G]
- Visited: {A, B, C, D, E, F, G}


#### Step 5:

Dequeue D

No unvisited children to add to queue --> no change to visited set

- Queue: [E, F, G]
- Visited: {A, B, C, D, E, F, G}


##### Step 6:

Dequeue E

No unvisited children to add to queue --> no change to visited set

- Queue: [F, G]
- Visited: {A, B, C, D, E, F, G}


#### Step 7:

Dequeue F

No unvisited children to add to queue --> no change to visited set

- Queue: [G]
- Visited: {A, B, C, D, E, F, G}


#### Step 8:

Dequeue G

No unvisited children to add to queue --> no change to visited set

- Queue: []
- Visited: {A, B, C, D, E, F, G}


#### Completed

Queue empty --> traversal complete
Nodes traversed in order: A, B, C, D, E, F, G

Algorithm visited nodes per level --> core principle of BFS


## DFS
Follows each branch as deep as possible then backtracks

Like exploring a maze --> choose specific path then follow until dead end/exit
If dead end --> go back & choose different path

Common uses:
- Solve single-solution puzzles
- Detect cycles in graph
- Find connected components in graph

Implement with recursion or stack data structure to track visited nodes (LIFO)

### How works:
1. Start at specific node
2. Mark as visited and add to stack
3. While stack not empty:
   - Current node popped from stack
   - Unvisited neighbours:
     - Marked as visited and added to stack

Limitation:
Not always shortest path between nodes in unweighted graph

### Example in Tree Graph


#### Step 1:
Start at root node (A)

Mark as visited and add to stack
- Stack: [A]
- Visited: {A}

#### Step 2:
Pop A from stack

Add univisited children to stack and mark as visited. Added in reverse order so B on top for LIFO
- Stack: [C, B]
- Visited: {A, B, C}


#### Step 3:
Pop B from stack

Add unvisited children to stack in rev order and mark as visited
- Stack: [C, E, D]
- Visited: {A, B, C, D, E}

#### Step 4:
Pop D from stack

No children to add to stack --> no change to visited set
- Stack: [C, E]
- Visited: {A, B, C, D, E}

#### Step 5:
Pop E from stack

No children to add to stack --> no change to visited set
- Stack: [C]
- Visited: {A, B, C, D, E}

#### Step 6:
Pop C from stack

Add unvisited children to stack in rev order and mark as visited
- Stack: [G, F]
- Visited: {A, B, C, D, E, F, G}

#### Step 7:
Pop F from stack

No children to add to stack --> no change to visited set
- Stack: [G]
- Visited: {A, B, C, D, E, F, G}

#### Step 8:
Pop G from stack

No children to add to stack --> no change to visited set
- Stack: []
- Visited: {A, B, C, D, E, F, G}

#### Completed
Stack empty --> traversal complete (all nodes visited)

Visited nodes in order: A, B, D, E, C, F, G

Traverses full paths then backtrack and find other paths

Could have been done with recursion:
- Func processes current nodes
- Calls itself for each unvisited neighbour
- Function call stack implicitely manages LIFO order
