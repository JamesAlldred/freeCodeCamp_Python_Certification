# Matrices and Adjacency Lists


## Two common ways to implement graphs in code

1. Adjacency Matrix --> for dense graphs with many edges
2. Adjacency List --> for sparse graphs with few edges (more real world)


## Adjacency Matrix

2D list

- Rows and cols --> graph vertices
- Values --> edges


IF you have matric `matrix`:

- Value at `matrix[i][j]` represents the edge from vertex `i` to vertex `j`
  - `i` is the row index
  - `j` is the column index
- If unweighted graph, the value is either 0 or 1
- If weighted graph, the value is the weight of the edge

Constant time `O(1)` to check if an edge exists between two vertices

- Only needs to find one value in 2D list
  Tradeoff: Quadratic space `O(V^2)`
- `V` is number of vertices
- If the graph is sparse, most of the values in the matrix will be 0, which is a waste of space
- Also inefficient to fin neighbors of a vertex, as you have to check all the values in the row or column `O(V)`

## Adjacency list

Array or dict that stores node neighbours


Two ways to implement:

1. Array --> each index rep node + list at that index stores neighbours
2. Dict --> each key rep node + value (a list) stores neighbours

More space efficient than adjacency matrix

- `O(V + E)` space, where `V` is number of vertices and `E` is number of edges
  Also efficient to ifnd neighbour nodes --> only needs to access list assoc to node

Tradeoff: Linear time `O(V)` to check if an edge exists between two vertices

- May have to search through log list if connected to all over nodes
