# Graphs in computer science
Data structures used to represent connection/relationship between objects/entities

Often used to model networks
- Social
- Transport
- Communication
- Recommendation systems

Very versatile

Often rep as collection of points connectred by lines: Nodes and edges
- Nodes/vertices: Rep objects/entities
- Edges: Connections

## Undirected Graphs

Edges dont have specific direction
Straight lines btwn nodes
- Connections bidirectional

## Directed Graphs

Edges have a specific direction
Lines end in arrows
If bidirectional connection is needed, two edges are used

## Vertex Labelled Graphs

Eaxh node has a label to identify, rep visually and store additional info

# Cyclic Graphs

Directed graph with at least one cycle
Ccle: Path to follow through edges that starts and ends at same node

# Edge Labelled Graphs

Edges associated with labels 

# Weighted Graphs

Edge labelled graph where labels rep values
- Values can be compared and used to do calcs
- Some edges have higher weight, others lower --> represent the "cost" of the edge

# Directed Acyclic Graphs

Directed graph with no cycles

# Disconnected Graphs

Two or more groups of nodes not connected by edges
