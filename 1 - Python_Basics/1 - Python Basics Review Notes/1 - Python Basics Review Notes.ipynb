```python
###Strings - Immutable

#Access characters
string = 'Hello'
print(string[3])

#Use \ if string contains ',"
#String concatenation (+, +=)

#f=strings
fstring = f'{string}, world!'
print(fstring)

#Slicing: [start:stop:step]
print(fstring[0:13:2])

#len() - num of characters
#in - whether character(s) ins string
#.upper/lower/title/capitalize()
#.isupper/islower() - True/False

#.strip - removes leading/trainling characters. Defaults whitespace
strip_str = " Spaces "
print(strip_str.strip())

#.replace()

#.split() - split into list with specified separator
split_str = "split-by-dashes"
print(split_str.split('-'))
#.join - into one string with separator
example_list = ['example', 'dashed', 'name']
join_str = ' '.join(example_list)
print(join_str)

#startswith/endswith(prefix/suffix)
#.find - index for first occurance. -1 if none

#.maketrans - creates 1:1 character mappings table. Often used w/ translate() - turns table to string
trans_table = str.maketrans('abc', '123')
print(trans_table)
result = 'abcabc'.translate(trans_table)
print(result)
```

```python
###Operations with intergers and floats

#+,-,*,/
#Modulo operator (%) - returns remainder when num/num
#Floor division operator (//) - num/num and rounds down to nearest int
#Exponential operator (**) - x^y

#float()
#int()
#round()
#abs() - returns absolute value
#pow() - x^y

#Augmented assignments - [operator]= --> applies operation to variable + stores result in same variable
example_int = 25
example_int *= 5
print(example_int)
```

```python
###Comparison operators

#== - checks if equal
#!= - checks if not equal
#> - checks if greater
#< - checks if less than
#>= - checks if greater or equal
#<= - checks if less than or equal
```

```python
### if, elif, else statements

#if - checks if true. If true, code runs
#elif - comes after if. Only runs if all prev false and own condition true
#else - runs if all else false

#nested if statements
condition1 = True
condition2 = 1

if condition1:
    if condition2 >= 1:
        print('Meets both conditions')
else:
    print('Doesn\'t meet both conditions')
```

```python
###Truthy and falsy values

##Falsy examples
#None
#None
#False
#Integer (0)
#Float (0.0)
#Empty strings ('')
##Non-zero, non-empty truthy

#bool() function - converts value to boolean equiv & returns True/False for Truthy/Falsy
print(bool(False))
print(bool(0)) 
print(bool(''))

print(bool(True))
print(bool(1))
print(bool('Hello'))

##Boolean Operators and Short-circuiting
#and - both True for truthy result
#or - one True for truthy result
#not - inverts boolean value. ALways returns True/False
```
