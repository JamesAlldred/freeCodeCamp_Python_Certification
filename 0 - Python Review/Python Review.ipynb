### Boolean Operators & Short-circuiting


Special operators to combine expressions for more complex decision making

- `and`: Takes two operands
  - Returns fist if falsy, otherwise returns second
  - Both must be truthy to result truthy value
- `or`: Takes two operands - Returns first if truthy, otherwise returns second - Only one must be truthy to result truthy value -`and` and `or` short-circuit: Python checks values L-->R and stops as soon as determines result
- `not`: Takes one operand and returns opposite boolean value
  - Always returns `True` or `False`

### `zip()`
Iterate over multiple iterables in parallel

````python
developers = ['Naomi', 'Dario', 'Jessica', 'Tom']
ids = [1, 2, 3, 4]

for name, dev_id in zip(developers, ids):
    print(f'Name: {name}')
    print(f'ID: {dev_id}')


"""
Result

Name: Naomi
ID: 1
Name: Dario
ID: 2
Name: Jessica
ID: 3
Name: Tom
ID: 4
"""
````

- `keys()`, `values()`, `items()`: Return view objects of keys, values, and key-value pairs respectively

Use `for` loop with `values()`, `keys()`, or `items()` to iterate over dictionary elements

- `enumerate()`: track index. Default start at 0

````python
products = {
    'Laptop': 990,
    'Smartphone': 600,
    'Tablet': 250,
    'Headphones': 70,
}
for product, price in enumerate(products.items()):
    print(product, price)
````

### Exception handling

#### `try...except`

Execute code block that might raise error
- `try` block: Where error might be
- `except`block:Takes specific exception & runs if specified error happens
    - Can chain `except` blocks together to handly multiple exceptions 

````python
list = [1, 2, 3, 4, 5]
try:
    print(list[5])
except IndexError:
    print("IndexError:Index out of range")
````

#### `else`, `finally`

Extend `try...except`
- If no exeption occurs, `else` block runs
- `finally` block always runs, regardless errors

````python
try:
    result = 100 / 4
except ZeroDivisionError:
    print("You cannot divide by zero!")  # This will not run
else:
    print(f"Result is {result}")  # Result is 25.0
finally:
    print("Execution complete!")  # Execution complete!
````

#### Exception object

Access exception itself for better debugging & error message printing
- `as` keyword

````python
try:
    value = int("This will raise an error")
except ValueError as e:
    print(f"Caught an error: {e}")
````

#### `raise` statement

Raise exception manually
- When condition met
- ALso useful for custom exeptions

````python
def divide(a, b):
    if b == 0:
        raise ZeroDivisionError("Dividing by zero is undefined.")
    return a / b
````

## Python Classes and Objects

- Class: blueprint for creating objects. Defines attributes and methods
    ```python
    class ClassName:
        def __init__(self, attribute1, attribute2):
            self.attribute1 = attribute1
            self.attribute2 = attribute2

        def method1(self):
            pass
    ```
- Object: instance of class. Can have unique attributes and methods
    - `object = ClassName(attribute1, attribute2)`
- Method: Functiond defined inside class that operate on object's attributes
    - `object.method1()`
    - Dunder (Magic) Methods: special methods that start & end with `__`
        - Used for internal Python operations, like `__init__`, `__str__`, `__repr__`, `__add__`, etc.
        - `__init__`: constructor method that initializes object's attributes when created
- Attributes: Variables associated with object. Can be accessed and modified using dot notation
    - Instance attributes: unique to each object. Defined in `__init__` method with `self`
    - Class attributes: defined directky in class. Shared by all instances of class

````python
# Class definition
class Car:
    # __init__(): Dunder method to initialize the object w/ instance attributes
    def __init__(self, color, model):
        self.color = color  # Instance attribute
        self.model = model  # self: refers to the instance of the class

    def describe(self):  # method
        return f"This car is a {self.color} {self.model}"


my_car_1 = Car("red", "Tesla Model S")  # object (instance of the class)
my_car_2 = Car("green", "Lamborghini Revuelto")

print(my_car_1.describe())  # Calling method using the dot notation

print(my_car_2.describe())  # Calling method using the dot notation
````
