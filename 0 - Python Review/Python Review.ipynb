## Python Classes and Objects

- Class: blueprint for creating objects. Defines attributes and methods
    ```python
    class ClassName:
        def __init__(self, attribute1, attribute2):
            self.attribute1 = attribute1
            self.attribute2 = attribute2

        def method1(self):
            pass
    ```
- Object: instance of class. Can have unique attributes and methods
    - `object = ClassName(attribute1, attribute2)`
- Method: Functiond defined inside class that operate on object's attributes
    - `object.method1()`
    - Dunder (Magic) Methods: special methods that start & end with `__`
        - Used for internal Python operations, like `__init__`, `__str__`, `__repr__`, `__add__`, etc.
        - `__init__`: constructor method that initializes object's attributes when created
- Attributes: Variables associated with object. Can be accessed and modified using dot notation
    - Instance attributes: unique to each object. Defined in `__init__` method with `self`
    - Class attributes: defined directky in class. Shared by all instances of class

````python
# Class definition
class Car:
    # __init__(): Dunder method to initialize the object w/ instance attributes
    def __init__(self, color, model):
        self.color = color  # Instance attribute
        self.model = model  # self: refers to the instance of the class

    def describe(self):  # method
        return f"This car is a {self.color} {self.model}"


my_car_1 = Car("red", "Tesla Model S")  # object (instance of the class)
my_car_2 = Car("green", "Lamborghini Revuelto")

print(my_car_1.describe())  # Calling method using the dot notation

print(my_car_2.describe())  # Calling method using the dot notation
````

### Encapsulation

Bundling of attributes & methods of onject into single unit (class)
- Hide internal state of object nehind set of public methods & attribs (doors)
    - Private attribs & methods behind --> control how data changes & who sees
- `_attribute`: Meant for internal use, convention --> doesn't enforce anything
- `__attribute`: Name mangling, prevents access from outside class
    - Python assigns name to `_ClassName__attribute` 

````python
class Wallet:
    def __init__(self, balance):
        self.__balance = balance  # Private attribute

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount  # Add to the balance safely

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount  # Remove from the balance safely


account = Wallet(500)
print(account.__balance)  # AttributeError: 'Wallet' object has no attribute '__balance'
````

### Getters and Setters

Methods that control access to class attributes
- Getters: Retrieve value
- Setters: Set values
- Properties: Connect getters and setters, allow access to data
    - Extra logic behind scenes when set/get/delete
    - Used instead of methods for readability --> access values w/ dot notation w/o ()

#### Getter, setter, deleter creation
- `@property`: Getter decorator
- `@<property_name>.setter`: Setter decorator
    - Don't assign to property name --> `RecursionError`
- `@<property_name>.deleter`: Deleter decorator

````python
class Circle:
    def __init__(self, radius):
        self.radius = radius  # Calling the setter

    @property
    def radius(self):  # A getter to get the radius
        return self._radius

    @radius.setter
    def radius(self, value):  # A setter to set the radius
        if value <= 0:
            raise ValueError("Radius must be positive")
        self._radius = value

    @radius.deleter
    def radius(self):  # A deleter to delete the radius
        print("Deleting radius...")
        del self._radius


my_circle = Circle(3)
print("Initial radius:", my_circle.radius)  # Initial radius: 3

my_circle.radius = 8
print("After modifying the radius:", my_circle.radius)  # After modifying the radius: 8

# del my_circle.radius  # Deleting radius...
# print("After deleting the radius:", my_circle.radius)  # AttributeError: 'Circle' object has no attribute '_radius'
````

#### Name Mangling (`__attribute`)

Python internally renames `__attribute` to `_ClassName__attribute` - Prevents access from outside class (accidental attrib/method overwriding during inheritance)
```python
class Parent:
    def __init__(self):
        self.__data = 'Parent data'

class Child(Parent):
    def __init__(self):
        super().__init__()
        self.__data = 'Child data'

c = Child()
print(c.__dict__) # {'_Parent__data': 'Parent data', '_Child__data': 'Child data'}
```

### Abstraction

Hiding implementation details, exposing only essential features
- Simplifies complex systems
- How Python implements:
    - `abc` module:
        - `ABC` class (ABstract Base Class)
            - Defines methods and properties that subclasses must implement
            - Can't be instantiated
        - `abstractmethod` decorator
            - Defines abstract method --> must be overwritten in subclasses


````python
from abc import ABC, abstractmethod


# Define an abstract base class
class AbstractClass(ABC):
    @abstractmethod
    def abstract_method(self):
        pass


# Concrete subclass that implements the abstract method
class ConcreteClassOne(AbstractClass):
    def abstract_method(self):
        print("Implementation in ConcreteClassOne")


# Another concrete subclass
class ConcreteClassTwo(AbstractClass):
    def abstract_method(self):
        print("Implementation in ConcreteClassTwo")
````

## Algorithms & Big O Notation

Algorithms: Set of unambiguous instructions to solve a problem or perform a task
- Finite number of steps
- Each step precise and unambiguous

Big O Notation: Describes worst-case performance/growth rate of algorithm as input size increases
- How resource usage grows with input size
- Ignores contant factors and lower order terms

### Time Complexity

- `O(1)`: Constant time --> same time regardless of input size
    - Algorithms that perform a fixed number of operations (accessing an element in a list)
- `O(log n)`: Logarithmic time --> time grows logarithmically w/ input size
    - Algorithms that reduce problem size by fraction (binary search)
- `O(n)`: Linear time --> time grows linearly w/ input size
    - Algorithms that iterate through all elements (linear search)
- `O(n log n)`: Log-linear time --> time grows log-linearly w/ input size
    - Efficient sorting algorithms (merge sort, quicksort)
- `O(n^2)`: Quadratic time --> time grows quadratically w/ input size
    - Algorithms that use nested loops (bubble sort, insertion sort)

### Space Complexity

- `O(1)`: Constant space --> Same memory regardless of input size
- `O(n)`: Linear space --> Memory grows linearly w/ input size
- `O(n^2)`: Quadratic space --> Memory grows quadratically w/ input size

### Searching and Sorting Algorithms

#### Searching algorithms

Search for target within list of items
1. Linear search: Check each item from beginning until target found
    - Found: Index returned
    - Not found: `-1` returned (invalid index)
    - Inefficient for large lists, but simple and works on unsorted lists
    - `O(n)` time complexity
    - `O(1)` space complexity
2. Binary search: Divide list in half, check if target in middle, repeat on half that contains target
    - Needs list in asc order
    - More efficient --> empty half ignored
    - Found: Index returned
    - Not found: Check left/right half --> continue dividing half with target until found or empty (`-1` returned)
    - `O(log n)` time complexity
    - `O(1)` space complexity

#### Sorting algorithms & divide and conquer

Divide and conquer: Break problem into smaller subproblems, solve each subproblem, combine solutions to solve original problem
- Recursion often used
- Merge sort: Divide list into sub-list until sub-lists only one element
- Merge back in sorted order
- `O(n log n)` time complexity
    - List continually halfed (log n) and merged back (n)
- `O(n)` space complexity

### Linked lists

#### Doubly linked lists

### Hash Maps and Sets

#### Maps & Hash Maps

Map (Abstract Data Type): Collection of key-value pairs
- Unique keys, repeatable values

Hash Map: Concrete implementation of map ADT using hashing
- Hash function: generates hash values for keys --> storage location in underlying array
- `O(1)`: Average case for insert, get, delete
- `O(n)`: Worst case (hash collisions)
- Python Dictionaries

````python
# Creating dictionaries
my_dictionary = {"A": 1, "B": 2, "C": 3}

# Alternative creation
my_dictionary = dict(A=1, B=2, C=3)

# Access and modify
value = my_dictionary["A"]  # 1
my_dictionary["A"] = 4  # Update value
del my_dictionary["A"]  # Remove key-value pair

# Check membership
"C" in my_dictionary

# Get keys, values, items
my_dictionary.keys()
my_dictionary.values()
my_dictionary.items()
````

#### Sets

Unordered collection of unique elements
- Immutable elements only --> hash values must remain constant
- `O(1)`: Average case for add, remove, check membership
- `O(n)`: Worst case (hash collisions)

````python
# Creating sets
numbers = {1, 2, 3, 4}
empty_set = set()  # Must use set(), not {}

# Add and remove elements
numbers.add(5)
numbers.remove(4)  # Raises KeyError if not found
numbers.discard(4)  # No error if not found

# Set operations
set_a = {1, 2, 3, 4}
set_b = {2, 3, 4, 5, 6}

# Union, intersection, difference, symmetric difference
set_a.union(set_b)  # or set_a | set_b
set_a.intersection(set_b)  # or set_a & set_b
set_a.difference(set_b)  # or set_a - set_b
set_a.symmetric_difference(set_b)  # or set_a ^ set_b

# Subset and superset checks
set_a.issubset(set_b)
set_a.issuperset(set_b)
set_a.isdisjoint(set_b)

# Membership testing
5 in numbers
````

#### Hash Collisions

2 different keys produce same hash value
- Collision resolution strategies:
    - Chaining: Each array index points to linked list of elements with same hash value
    - Open addressing: Search for next available index using predefined sequence

### When to use each Data Structure

- Lists: When you need ordered, indexed access and don't know size in advance
- Stacks: For LIFO operations (undo functionality, expression evaluation, backtracking)
- Queues: For FIFO operations (task scheduling, breadth-first search)
- Linked Lists: When frequent insertion/deletion at beginning, unknown size, no random access needed
- Hash Maps: For fast key-value lookups, counting occurrences, caching
- Sets: For uniqueness checking, mathematical set operations, removing duplicates

- Each node can connect to multiple nodes --> network
- Types:
    - Directed: edges have a direction (from one node to another), often represented with straight lines and arrows.
    - Undirected: edges have no direction, represented with simple lines.
    - Vertex: each node is associated with a label or identifier.
    - Cyclic: contains cycles (a path that starts and ends at the same node).
    - Acyclic (DAG): does not contain cycles.
    - Edge labeled: each edge has a label usually drawn next to corresponding edge.
    - Weighted: edges have weights (values) associated with them, that can be used to perform arithmetic operations.
    - Disconnected: contains two or more nodes that are not connected by any edges.
- Applications: Maps, networks, recommendation systems, dependency resolution...

#### Graph Traversals


Visit every node
1. Breadth-First Search (BFS):
    - Uses queue
    - Explores level by level
    - Finds shortest path in unweighted graphs
2. Depth-First Search (DFS):
    - Uses stack or recursion
    - Explores branch fully before backtracking
    - Useful for cucle detection, path finding

#### Graph Representations

1. Adjacency List: Each node has a list of its neighbors
    - Space efficient for sparse graphs
    - Easy to iterate over neighbors
2. Adjacency Matrix: 2D array where rows and columns represent nodes, and values indicate presence/absence of edges
    - Space intensive for large graphs
    - Easy to check if edge exists between two nodes

#### Trees

Acyclic connected graph
- No loops or cycles (paths where start and end node are the same)
- Connected: All nodes connected by edges

Most common types:
1. Binary Tree: Each node has at most 2 children (left and right)
2. Binary Search Tree (BST): Binary tree where left child < parent < right child

##### Tries

Tries/Prefix Trees: Store sets of strings in a tree structure
- Each node rep character
- Shared prefixed only stored once --> memory efficient for autocomplete, spell check
- Search, insertion operations: `O(L)`
    - L = length of string

##### Heaps

Specialised tree-based data structure w/ special heap property
- Heap property: determines child-parent node relationship
1. Max-Heap:
    - Parent value >= child values
    - Largest element @ root
2. Min-Heap:
    - Parent value <= child values
    - Smallest element @ root

###### `heapq` module

````python
import heapq

# Create empty heap
my_heap = []

# Insert elements
heapq.heappush(my_heap, 9)
heapq.heappush(my_heap, 3)
heapq.heappush(my_heap, 5)
print(my_heap)  # [3, 9, 5]

# Remove smallest element
print(heapq.heappop(my_heap))  # 3
print(my_heap)  # [5, 9]

# Push + Pop in one step
print(heapq.heappushpop(my_heap, 7))  # 5
print(my_heap)  # [7, 9]

# Transform list into heap
nums = [5, 7, 3, 1]
heapq.heapify(nums)
````

###### Using Priority Queues

````python
my_heap = []
heapq.heappush(my_heap, (3, "A"))
heapq.heappush(my_heap, (2, "B"))
heapq.heappush(my_heap, (1, "C"))

# Removes lowest number = highest priority
print(heapq.heappop(my_heap))  # (1, "C")
````

## Dynamic Programming

Algorithmmic technique to solve complex problems by breaking down to simpler subproblems & storing results of subproblems to avoid redundant computations
- Needs:
    - Overlapping subproblems: Problem can be broken down into smaller subproblems that are solved multiple times
    - Optimal substructure: Optimal solution to problem can be constructed from optimal solutions to its subproblems

### Dynamic Programming Solutions

#### Memoization (Top-down)

Store results of expensive function calls & returns cached result when same inputs occur again

````python
def climb_stairs_memo(n, memo={}):
    """Dynamic programming with memoization"""
    # Check if we've already calculated this value
    if n in memo:
        return memo[n]  # Return cached result - O(1) lookup!

    # Base cases
    if n <= 2:
        return n

    # Calculate once and store in memo for future use
    memo[n] = climb_stairs_memo(n - 1, memo) + climb_stairs_memo(n - 2, memo)
    return memo[n]
````

#### Tabulation (Bottom-up)

Build solution from ground up, filling table with subproblem solutions

````python
def climb_stairs_tabulation(n):
    """Dynamic programming with tabulation"""
    if n <= 2:
        return n

    # Create array to store results for all steps from 0 to n
    dp = [0] * (n + 1)
    dp[1] = 1  # 1 way to reach step 1
    dp[2] = 2  # 2 ways to reach step 2

    # Build up the solution iteratively
    for i in range(3, n + 1):
        # Ways to reach step i = ways to reach (i-1) + ways to reach (i-2)
        dp[i] = dp[i - 1] + dp[i - 2]

    return dp[n]
````

### Dynamic Programming applications

- Route Optimization: GPS systems use dynamic programming algorithms to find shortest paths between locations.
- Text Processing: Spell checkers and autocomplete features often rely on dynamic programming to calculate edit distances between words.
- Financial Modeling: Investment strategies and portfolio optimization frequently employ dynamic programming techniques.
- Resource Allocation: The knapsack problem and its variants appear in scheduling, budgeting, and resource management.

### When to use dynamic programming

- Overlapping subproblems
- Optimal substructure
- Naive recusrive solution too many calcs
- Need to optimise for time over space
