## Truthy and Falsy Values


Every value has inherent boolean value. `True` or `False`.
Falsy examples:

- `None`
- `False`
- Integer `0`
- Float `0.0`
- Empty strings `''`


### `bool()` function: Check if a value is `True` or `False`


Converts value to boolean equiv

- Returns `True` for truthy values and `False` for falsy values


````python
print(bool(False)) # False
print(bool(0))  # False
print(bool('')) # False

print(bool(True)) # True
print(bool(1)) # True
print(bool('Hello')) # True
````

### Boolean Operators & Short-circuiting


Special operators to combine expressions for more complex decision making

- `and`: Takes two operands
  - Returns fist if falsy, otherwise returns second
  - Both must be truthy to result truthy value
- `or`: Takes two operands - Returns first if truthy, otherwise returns second - Only one must be truthy to result truthy value -`and` and `or` short-circuit: Python checks values L-->R and stops as soon as determines result
- `not`: Takes one operand and returns opposite boolean value
  - Always returns `True` or `False`

#### Unpacking values from list
Assign values from list to new variables

````python
cricketer = ['Root', 'RHB', 'England']
name, role, team = cricketer
print(name) # Root

# ValueError: too many values to unpack
# Collect remaining items with `*`
cricketer_1 = ['Root', 'RHB', 'England', 'Captain']
name, role, *rest = cricketer_1
print(rest) # ['England', 'Captain']
````

#### Slicing lists (`:`)

````python
sports = ['Cricket', 'Football', 'Tennis', 'Basketball', 'Hockey']
sports[1:4] # ['Football', 'Tennis', 'Basketball'

# Step
sports[::2] # ['Cricket', 'Tennis', 'Hockey']
````

## `enumerate()`, `zip()`


### `enumerate()`
Iterate over seq & track index of each item

- Takes iterable as arg
- Returns `enumerate` object (index + value of each item)
- Can be used outside for loop
- Optional `start` arg to specify starting index (default 0)

### `zip()`
Iterate over multiple iterables in parallel

````python
developers = ['Naomi', 'Dario', 'Jessica', 'Tom']
ids = [1, 2, 3, 4]

for name, dev_id in zip(developers, ids):
    print(f'Name: {name}')
    print(f'ID: {dev_id}')


"""
Result

Name: Naomi
ID: 1
Name: Dario
ID: 2
Name: Jessica
ID: 3
Name: Tom
ID: 4
"""
````

## List Comprehensions

Create new list in single line

## Iterable functions

- `filter()`: filter elements from iterable based on condition
    - Returns iterator that contains only satisfying elements
- `map()`: apply function to each item in iterable
    - Returns new iterable with results
- `sum()`: Returns sum of all items in iterable
    - Optional `start` arg (value not index)

## Lambda functions

Concise way to create function without name
- Often used as arg to another function

````python
numbers = [1, 2, 3, 4, 5]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print(even_numbers)  # [2, 4]
````

## Dictionaries

Built in data structure that store KV pairs
- Keys need to be immutable
- `dict()` or `{}` to create

### Dictionary methods

- `get()`: Retireves value assoc w/ key
    - Similar to using `[]`, but can set default value `dictionary.get(key, default_value)` to avoid error if key doesn't exist
- `keys()`, `values()`, `items()`: Return view objects of keys, values, and key-value pairs respectively
- `clear()`: Remove all items from dictionary
- `pop(key)`: Remove key-value pair and return value. Raises KeyError if key doesn't exist
- `popitem()`: Remove last inserted item
- `update()`: Update dictionary with key-value pairs from another dictionary
    - Overwritten if equal
    - New keys added as new KV pairs

### Looping over a dictionary

Use `for` loop with `values()`, `keys()`, or `items()` to iterate over dictionary elements
- To store key and value in separate loop variables, separate with comma (key, value)
- `enumerate()`: track index. Default start at 0

````python
products = {
    'Laptop': 990,
    'Smartphone': 600,
    'Tablet': 250,
    'Headphones': 70,
}
for product, price in enumerate(products.items()):
    print(product, price)
````

## Sets

Built in data structure
- No duplicate values
- Mutable
- Unordered --> can't use indices or keys to access
- Only contain immutable data types
- Define with `set()` or `{}`
    - Have to use `set()` to create empty set, `{}` creates empty dictionary

### Set methods

- `add(item)`: Add element to set
- `remove(item)`: Remove element from set. Raises KeyError if not found
- `discard(item)`: Remove element from set. Does not raise error if not found
- `clear()`: Remove all elements from set

### Mathematical set operations

- `issubset()`, `issuperset()`: Check if one set is subset or superset of another
- `isdisjoint()`: Check if two sets have no elements in common
- Union operator `|`: Combines elements from two sets, removing duplicates
- Intersection operator `&`: Returns elements that are in both sets
- Difference operator `-`: Returns elements that are in the first set but not in the second set
- Symmetric difference operator `^`: Returns elements that are in either set, but not in both sets
- `in` operator: Check if element exists in set. Returns bool
