## MediaError Class

- Inherits from the inbuilt `Exception` class to define custom MediaError message
- Passes custom error message to parent class & stores specific media item that triggered exception (`self.obj`) for debugging


```python
class MediaError(Exception):
    """Custom exception for media-related errors.""" # Docstring: Labels what class does
    
    def __init__(self, message, obj): # Message: Error message, obj: Object related to the error
        super().__init__(message) # super(): Preserve parent class behavior
        self.obj = obj # Stores error related object
```

## Movie Class (parent)

- Makes structure for movie data to be entered
- Validates data entered to specs
- Defines print behaviour


```python
class Movie: 
    """Parent class representing a movie."""
    
    def __init__(self, title, year, director, duration):
        if not title.strip(): # strip(): Removes leading and trailing whitespace
            raise ValueError('Title cannot be empty')
        if year < 1895:
            raise ValueError('Year must be 1895 or later')
        if not director.strip():
            raise ValueError('Director cannot be empty')
        if duration <= 0:
            raise ValueError('Duration must be positive')
        
        self.title = title
        self.year = year
        self.director = director
        self.duration = duration
    
    def __str__(self): # String representation of the Movie object. Print behavior of the object when called.
        return f'{self.title} ({self.year}) - {self.duration} min, {self.director}'
```

### TVSeries Class (child)

- Similar structure for movie data
  - `super()`: Preserves parent class behaviour
  - `super().__init__()`: Calls parent (Movie) class constructor to initialize common attributes
- Validates data entered to specs
- Defines print behaviour


```python
class TVSeries(Movie):
    """Child class representing an entire TV series."""
    
    def __init__(self, title, year, director, duration, seasons, total_episodes): 
        super().__init__(title, year, director, duration) # Calls the parent class constructor to initialize common attributes.
        
        if seasons < 1:
            raise ValueError('Seasons must be 1 or greater')
        if total_episodes < 1:
            raise ValueError('Total episodes must be 1 or greater')
        
        self.seasons = seasons
        self.total_episodes = total_episodes
    
    def __str__(self):
        return f'{self.title} ({self.year}) - {self.seasons} seasons, {self.total_episodes} episodes, {self.duration} min avg, {self.director}'
```

## MediaCatalogue Class: Blueprint to store the media items

- Starts as empty list
- Function to add media items (validates --> if MediaError the MediaError class stores the problem object)
- Uses loops to split movies & TV series from list
- Prints formatted string representation
  - Calls earlier loops to split
  - Gives number of items in list
  - Header
  - Adds each movie/series to result with its index (starting at 1)


```python
class MediaCatalogue:
    """A catalogue that can store different types of media items."""
    
    def __init__(self):
        self.items = [] # Empty list to store media items
    
    def add(self, media_item): # Adds media item to catalogue
        if not isinstance(media_item, Movie):
            raise MediaError('Only Movie or TVSeries instances can be added', media_item) # Checks for MediaError
        self.items.append(media_item) 
    
    def get_movies(self): # Loops through items list and returns movies/tvseries
        return [item for item in self.items if type(item) is Movie]
    
    def get_tv_series(self):
        return [item for item in self.items if isinstance(item, TVSeries)]
    
    def __str__(self): # String representation of the MediaCatalogue object. Print behavior of the object when called.
        if not self.items:
            return 'Media Catalogue (empty)'
        
        movies = self.get_movies()  # No def because in def __str__()
        series = self.get_tv_series()
        
        result = f'Media Catalogue ({len(self.items)} items):\n\n' # Number of items
        if movies:
            result += '=== MOVIES ===\n' 
            for i, movie in enumerate(movies, 1): # Enumerate to get index + 1
                result += f'{i}. {movie}\n' # Add each movie to the result 
        
        if series:
            result += '=== TV SERIES ===\n'
            for i, series in enumerate(series, 1):
                result += f'{i}. {series}\n'        
        
        return result
```

## Catalogue instances

- Enter & store data
- `try:`: Custom error messages if error
- `Except`: Runs if error message
  - `e`: Prints error
  - Uses MediaError to pinpoint error


```python
catalogue = MediaCatalogue() #No arguments are needed bc MediaCatalogue class initializes an empty list to store data when instance created
try:
    movie1 = Movie('The Matrix', 1999, 'The Wachowskis', 136)
    catalogue.add(movie1) # Calls add method 
    movie2 = Movie('Inception', 2010, 'Christopher Nolan', 148)
    catalogue.add(movie2)
    
    series1 = TVSeries('Scrubs', 2001, 'Bill Lawrence', 24, 9, 182)
    catalogue.add(series1)
    series2 = TVSeries('Breaking Bad', 2008, 'Vince Gilligan', 47, 5, 62)
    catalogue.add(series2)

    print(catalogue)
except ValueError as e: # e: exception object. Gives info about the error
    print(f'Validation Error: {e}')
except MediaError as e:
    print(f'Media Error: {e}')
    print(f'Unable to add {e.obj}: {type(e.obj)}') # Returns the object that caused the error and its type
```
