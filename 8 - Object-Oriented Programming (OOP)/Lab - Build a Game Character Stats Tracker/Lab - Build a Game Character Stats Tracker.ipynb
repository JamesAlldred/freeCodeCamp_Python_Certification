## Class & property setup


````python
class GameCharacter: # Make class
    def __init__(self, name): # Uses _ to show private instance attributes
        self._name = name
        self._health = 100
        self._mana = 50
        self._level = 1
    
    @property # Getter decorator for name attribute. No setter as is read only
    def name(self):
        return self._name
    
    @property
    def health(self):
        return self._health

    @health.setter # Setter decorator for health attribute
    def health(self, new_health):
        if new_health < 0:
            self._health = 0 # Pins health to > 0 
        elif new_health > 100:
            self._health = 100 # Pins health to < 100
        else:
            self._health = new_health # Allows user to set new health value 
    
    @property
    def mana(self):
        return self._mana
    
    @mana.setter
    def mana(self, new_mana):
        if new_mana < 0:
            self._mana = 0
        elif new_mana > 50:
            self._mana = 50
        else:
            self._mana = new_mana
    
    @property # Getter decorator for level attribute. No setter as will be edited by level_up() method
    def level(self):
        return self._level

    def level_up(self): # Method to level up character. Increases level by 1, resets health & mana to max, and prints message.
        self._level = self._level + 1
        self.health = 100
        self.mana = 50
        print(f"{self._name} leveled up to {self._level}!")
    
    def __str__(self): # __str__() is a dunder method that returns a string representation of the object. It is called when you use the print() function on an instance of the class.
        return f"Name: {self._name}\nLevel: {self._level}\nHealth: {self._health}\nMana: {self._mana}" # Returns levelled up stats
````

## Usage


### Create object/instance of class - `__str__()` method gives stats when print() called


````python
character = GameCharacter('Kaos')
print(character)
````

### Edit stats using the `.health` and `.mana` properties


````python
character.health -= 50
character.mana -=25
print(character)
````

### Level up with the `level_up()` method


````python
character.level_up()
print(character)
````
