# Abstraction: hide complex details & show only essential features


## Uses the `abc` module

- `ABC` class (Abstract Base Class): Class meant to inherit from
  - Can't create direct objects from it
  - Defines common interface of methods & properties that subclasses must implement
- `@abstractmethod` decorator: Declares abstract method
  - May have no implementation or basic default
  - Any subclass must override it to be concrete & instantiable
- Basic Syntax:

  ```python
  from abc import ABC, abstractmethod

  # Define an abstract base class
  class AbstractClass(ABC):
    @abstractmethod
    def abstract_method(self):
      pass

  # Concrete subclass that implements abstract method (could have multiple subclasses)
  class ConcreteClassOne(AbstractClass):
    def abstract_method(self): # Overrides
      print('Implementation in ConcreteClassOne)
  ```


## Example

- `TalkingToy` abstract base class --> blueprint for any speaking toy
- Subclasses override `speak()` in their own way
- Create instances of the subclasses indirectly with list & loop
  - Can't make instance of class or subclass without overriding first


````python
from abc import ABC, abstractmethod

# The blueprint for any toy that can speak
class TalkingToy(ABC):
   def __init__(self, name):
       self.name = name
   @abstractmethod
   def speak(self):
       pass

class RobotToy(TalkingToy):
   def speak(self):
       print(f'{self.name} says beep boop! I am a robot!')

class TeddyBearToy(TalkingToy):
   def speak(self):
       print(f"{self.name} says hug me! I'm cuddly!")

class DinosaurToy(TalkingToy):
   def speak(self):
       print(f'{self.name} says ROOOOAR!')

# Create toys
rusty = RobotToy('Rusty')
fluffy = TeddyBearToy('Fluffy')
rex = DinosaurToy('Rex')

toys = [rusty, fluffy, rex]
for toy in toys:
   toy.speak()
````
