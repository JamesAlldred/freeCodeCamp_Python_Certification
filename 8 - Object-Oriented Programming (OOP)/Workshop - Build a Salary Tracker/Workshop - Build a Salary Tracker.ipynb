# Workshop - Build a Salary Tracker


## Context & Definitions:

- OOP: Programming language where devs treat everything in code like real-world object
- Class: Bluepring for objects.
- Objects in class: Attriubute that define data & methods that define behaviours

- Encapsulation: bundling of attributes & methods of object into single unit (class)
  - Hide internal state of object behind public methods & attributes (doors). Private methods & attributes behind doors that controls how data changes & who sees it
  - `_attribute`, `_method`: Only for internal use (convention not function)
  - `__attribute`, `__method`: Prevents being accessed outside of class (actually private)

- Getters & setters: Control how attributes of class accessed & modified.
  - Getters: Retrieve value
  - Setters: Set value

  - Properties: Connect getters & setters and allow access to data
    - Define like methods but work like attributes - access with dot notation (`.property`)
    - Run extra logic behind scenes when set/get/delete values with them --> access/manip data w/in objects
    - Decorator: Func that nodifies funcs of other funcs/classes w/o changing code
    - Create property: def method & place `@property` decorator above
  - Making getter:
    ```python
    @property
    def radius(self):
      return self._radius # Getter to get radius
    ```
  - Making setter:
    - Inside `__init__` so setter runs during object creation --> invalid vaalues caught immediately
    ```python
    @method.setter
    def radius(self, value): # Setter to set radius
        if value <= 0:
            raise ValueError('Radius must be positive')
        self._radius = value
    ```
  - Once defined, call with attribute syntax: `my_circle.radius` (getter), `my_circle.radius = 4` (setter)

- Deleter: Controls how attribute deleted
  - Runs custom logic when use `del` statement on property
  - Making deleter:

  ```python
  @radius.deleter
  def radius(self):
    print("Deleting radius...")
    del self.radius
    '''

  ```

- Summary:
  - Encapsulation: bundling of attribs & methods into class
    - Private: `_` labels, `__` actually stops access outside of class
  - Getters: Retrieve/compute value
  - Setters: Modify & validate value
  - Properties: Tie getters & setters, dot notation
  - Deleters: Define what happens when attrib deleted


````python
class Employee: # Make the class
    _base_salaries = { # (_) is used to show private attribute + shouldn't be accessed directly from outside class.
        'trainee': 1000,
        'junior': 2000,
        'mid-level': 3000,
        'senior': 4000,
    }                  # Dictionary is used to store the base salaries for each level. Class attribute (shared among all instances of the class)

    def __init__(self, name, level): # Three parameters: self, name, level. self - instance of class. name + level - instance attributes
        self.name = name
        self.level = level
        self.salary = Employee._base_salaries[level] # Set the salary based on the level using the class attribute _base_salaries.

    def __str__(self): # String representation of the object. When you print an instance of the class, this method is called.
        return f'{self.name}: {self.level}' # Outputs as "name: level".

    def __repr__(self): # String representation of the object for debugging purposes. When you call repr() on an instance of the class, this method is called.
        return f"Employee('{self.name}', '{self.level}')" # Outputs as "Employee('name', 'level')".

    @property # Getter decorator
    def name(self):
        return self._name # Access with self._name instead of self.name to avoid infinite recursion.

    @name.setter # Setter decorator
    def name(self, new_name): # Change name & validate new_name
        if not isinstance(new_name, str):
            raise TypeError("'name' must be a string.")
        self._name = new_name # Sets the new name as the name. _name to avoid infinite recursion
        print(f"'name' updated to '{self.name}'.")

    @property 
    def level(self):
        return self._level

    @level.setter
    def level(self, new_level): # Change level & validate new_level
        if not isinstance(new_level, str):
            raise TypeError("'level' must be a string.")
        if new_level not in Employee._base_salaries:
            raise ValueError(f"Invalid value '{new_level}' for 'level' attribute.")
        if hasattr(self, '_level') and new_level == self.level: # hasattr() checks if instance same as current level. If yes, raise ValueError.
            raise ValueError(f"'{self.level}' is already the selected level.")
        if hasattr(self, '_level') and Employee._base_salaries[new_level] < Employee._base_salaries[self.level]:
            raise ValueError("Cannot change to lower level.")
        print(f"'{self.name}' promoted to '{new_level}'.")
        self.salary = Employee._base_salaries[new_level] # Update salary based on new level using the class attribute _base_salaries.
        self._level = new_level # Sets the new level as the level. _level to avoid infinite recursion.

    @property
    def salary(self):
        return self._salary

    @salary.setter
    def salary(self, new_salary): # Change salary & validate new_salary
        if not isinstance(new_salary, (int, float)):
            raise TypeError("'salary' must be a number.")
        if hasattr(self, '_level') and new_salary < Employee._base_salaries[self.level]:
            raise ValueError(f'Salary must be higher than minimum salary ${Employee._base_salaries[self.level]}.')
        self._salary = new_salary # Sets the new salary as the salary. _salary to avoid infinite recursion.
        print(f'Salary updated to ${self.salary}.')

````

### Object & property calls


````python
charlie_brown = Employee('Charlie Brown', 'trainee') # Creates instance/object with name & level. Calls __init__() method.
print(charlie_brown) # Calls __str__() method. Outputs as "name: level". If __str__() method is not defined, it will call __repr__() method.
print(f'Base salary: ${charlie_brown.salary}')
charlie_brown.level = 'junior' # Calls level.setter method. Updates level & salary. Outputs as "'name' promoted to 'new_level'." & "Salary updated to $new_salary."
````
