# Workshop - Build a Discount Calculator


```python
from abc import ABC, abstractmethod
```

```python
class Product: # Product with name and price
    def __init__(self, name: str, price: float) -> None: # Type hints: indicate expected data types
        self.name = name
        self.price = price

    def __str__(self) -> str:
        return f'{self.name} - ${self.price}' # Template for printing
```

```python
class DiscountStrategy(ABC): # Inherits from ABC: Abstract class for discount strategies
    @abstractmethod # Decorator: indicates that the method must be implemented in subclasses
    def is_applicable(self, product: Product, user_tier: str) -> bool: # Check if the discount is applicable to product & tier
        pass

    @abstractmethod
    def apply_discount(self, product: Product) -> float: # Apply discount
        pass
```

```python
class PercentageDiscount(DiscountStrategy): # First discount strategy
    def __init__(self, percent: int) -> None:
        self.percent = percent

    def is_applicable(self, product: Product, user_tier: str) -> bool:
        return self.percent <= 70 # Example condition: discount is applicable if percent is less than or equal to 70

    def apply_discount(self, product: Product) -> float: # Apply the percentage discount
        return product.price * (1 - self.percent / 100)
```

```python
class FixedAmountDiscount(DiscountStrategy): # Second discount strategy
    def __init__(self, amount: int) -> None:
        self.amount = amount

    def is_applicable(self, product: Product, user_tier: str) -> bool:
        return product.price * 0.9 > self.amount # Example condition: discount is applicable if product price is greater than 10% of the amount

    def apply_discount(self, product: Product) -> float:
        return product.price - self.amount
```

```python
class PremiumUserDiscount(DiscountStrategy): # Third discount strategy
    def is_applicable(self, product: Product, user_tier: str) -> bool:
        return user_tier.lower() == 'premium' # Example condition: discount is applicable if user tier is 'premium'

    def apply_discount(self, product: Product) -> float:
        return product.price * 0.8  
```

```python
class DiscountEngine: # Engine to calculate best price based on strategies
    def __init__(self, strategies: list[DiscountStrategy]) -> None: # Expected to list of DiscountStrategy (abstract class)instances
        self.strategies = strategies

    def calculate_best_price(self, product: Product, user_tier: str) -> float:
        prices = [product.price] # Initialize prices list with original product price

        for strategy in self.strategies: # Iterate through each discount strategy
            if strategy.is_applicable(product, user_tier):
                discounted = strategy.apply_discount(product)
                prices.append(discounted)
        return min(prices) 
```

```python
if __name__ == '__main__': # This block will only run if the script is executed directly, not when imported as a module
    product = Product('Wireless Mouse', 50.0)
    user_tier = 'Premium'

    strategies = [
        PercentageDiscount(10),
        FixedAmountDiscount(5),
        PremiumUserDiscount()
    ] # List for the strategies parameter of DiscountEngine

    engine = DiscountEngine(strategies) # Instance
    best_price = engine.calculate_best_price(product, user_tier) # Call best_price method
    print(f"Best price for {product.name} for {user_tier} user: ${best_price:.2f}") 
```
