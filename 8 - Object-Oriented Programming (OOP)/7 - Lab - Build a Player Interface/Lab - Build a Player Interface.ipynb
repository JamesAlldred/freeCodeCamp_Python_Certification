# Build a player interface


## Code setup

- Imports
- Parent class
  - Methods
- Subclass


```python
from abc import ABC, abstractmethod # Modules to make abstract classes and methods: Must be implemented in subclasses
import random # Make randon moves

class Player(ABC):# Parent class
    def __init__(self):
        self.moves = [] # Empty list: Will be filled with moves in subclasses
        self.position = (0,0) # Starting position
        self.path = [self.position] # List to keep track of the path taken by the player
    
    def make_move(self): # Moving method
        random_move = random.choice(self.moves) # Randomly choose a move from the list of possible moves
        new_x = self.position[0] + random_move[0] # Self.position[0/1]: Gets x/y position (x, y) --> 0/1 is first/second in tuple
        new_y = self.position[1] + random_move[1] # random_move[0/1]: Gets x/y move. Gets moves from self.moves list
        self.position = (new_x, new_y)
        self.path.append(self.position)
        return self.position
    
    @abstractmethod
    def level_up(self): # Abstract method: Must be implemented in subclasses. 
        pass

class Pawn(Player):
    def __init__(self):
        super().__init__() # Call the constructor of the parent class 
        self.moves = [(0, 1), (0, -1), (-1, 0), (1, 0)] # moves for random_move
    def level_up(self):
        self.moves.extend([(1, 1), (1, -1), (-1, 1), (-1, -1)]) # moves for level_up
```

## Using code

- Create Pawn() object & call methods


```python
pawn = Pawn()

print(pawn.position)
print(pawn.make_move())
print(pawn.position)
print(pawn.path)

pawn.level_up()
print(pawn.make_move())
print(pawn.position)
print(pawn.path)
```
