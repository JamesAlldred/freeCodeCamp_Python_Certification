# OOP Review


## Object-Oriented Programming

- Programming style where everything in code treated as real world object
- 4 key principles:
  1. Encapsulation
  2. Inheritance
  3. Polymorphism
  4. Abstraction
- Classes: Blueprint for objects
  - Every object created from class:
    - Attributes: Define data
    - Methods: Determine behaviour


## Encapsulation

- Bundling of attributes & methods of object into single unit
  - Hide internal state of object behind public methods & attributes (doors)
  - Behind doors --> private attributes & methods that control how the data changes & who sees it


### Encapsulation example: Track wallet balance

- Allow deposit & withdrawal
- Don't allow balance to be directly changed

- `_`: Meant for internal use
- `__`: Prevents from being accesssed outside of class (name mangling)
  - Python renames the attribute (`__attribute` --> `_ClassName__attribute`)


````python
class Wallet:
    def __init__(self, balance):
        self.__balance = balance

    @property
    def balance(self):
        return self.__balance

    def deposit(self, amount):
        if amount > 0:self.__balance += amount
    
    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
    
account = Wallet(500)
print(account.__balance) # AttributeError: __balance private attribute
````

### Encapsulation : Getters & Setters

- Methods to control how attribute accessed & modified
  - Getters: Retrieve values
  - Setters: Set values
- Properties: Connect getters & setters (allow access to data). Run extra logic
  - Used over methods for better readability & cleaner code
  - Dot notation like attributes (`.property`)

- Creation:
  - Getter: `@property` decorator
  - Setter: `@<property_name>.setter` decorator

- Can't assign t property name itself --> `RecursionError`
  - Use sepatate internal name (`_`)

- Deleter: Control how value deleted. Runs custom logic
  - `del <property_name>`
  - `@<property_name>.deleter` decorator


## Inheritance

- Child class uses attributes & methods of parent class
  - Promotes code reuse
  - Clear hierarchy
  - Customise behaviour w/o rewriting
- Basic Syntax:

```python
class Parent:
    pass # Parent attributes & methods
class Child(Parent):
    pass # Inherits/extends/overrides Parent attributes & methods
```

- Single inheritance: 1 parent class
- Multiple inheritance: multiple parents

  ```python
  class Grandchild(Parent, Child):
      pass # Inherits from both, combines or overrides
  ```

- `super()`: Call method from parent class
  - When class have different implementation or extends method w/o duplication


## Polymorphism

- Different classes use same method name, but each class implements differently
- Basic syntax:

```python
class A:
   def action(self): ...

class B:
   def action(self): ...

class C:
   def action(self): ...

Class().method()  # Works for A, B, or C
```


### Name mangling

- Internal Python process where `__attribute` is renamed to `_ClassName__attribute`
  - Prevents accidental attribute & method overwriding during inheritance


#### Name mangling example


````python
class Parent:
    def __init__(self):
        self.__data = 'Parent data'

class Child(Parent):
    def __init__(self):
        super().__init__()
        self.__data = 'Child data'

c = Child()
print(c.__dict__) # {'_Parent__data': 'Parent data', '_Child__data': 'Child data'}
````

## Abstaction

- Complex details hidden, only essential shown
- Uses `abc` module
  - `ABC` class (Abstract Base Class): Defines methods & properties subclasses must implement.
    - Can't be instantiated
    - Must be overwriddden in subclasses
  - `@abstractmethod` decorator

- Basic syntax:

```python
from abc import ABC, abstractmethod

# Define an abstract base class
class AbstractClass(ABC):
    @abstractmethod
    def abstract_method(self):
        pass

# Concrete subclass that implements the abstract method
class ConcreteClassOne(AbstractClass):
    def abstract_method(self):
        print('Implementation in ConcreteClassOne')

# Another concrete subclass
class ConcreteClassTwo(AbstractClass):
    def abstract_method(self):
        print('Implementation in ConcreteClassTwo')
```
