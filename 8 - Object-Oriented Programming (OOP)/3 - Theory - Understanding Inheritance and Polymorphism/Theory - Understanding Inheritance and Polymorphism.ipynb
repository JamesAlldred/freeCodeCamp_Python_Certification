## Inheritance (Code Reuse)

- Subclass/child class can use attributes & methods of base/parent class --> Reuse code without rewriting
  - Customise by extending existing methods or overriding in child class
- Inheritance basic syntax:

  ```python
  class Parent:
  # Parent attributes and methods

  class Child(Parent):
  # Child inherits, extends, and/or overrides where necessary
  ```


### Single inheritance: have to pass the parent to the child

- Passing parent to child, overriding or extending with super() example:

  ```python
  class Cricketer: # Parent ckass
      def __init__(self, name):
          self.name = name

      def action(self): # Method
          return f"{self.name} plays cricket"

  class Batter(Cricketer): # Child class
      shot = 'cover drive'

      def action(self): # Override action() to use shot class variable
          return f'{self.name} plays a {self.shot}'

      def action(self): # Keep return value of action & adds shot after using super()
          base = super().action()
          return f'{base}, and {self.name} has a nice {self.shot}'

  root = Batter('Root')
  print(root.action()) # Ouput: "Root plays a cover drive" OR "Root plays cricker, and Root has a nice cover drive"
  ```


### Multiple inheritance: Child class inherits from more than one parent

- Basic syntax:

  ```python
  class Parent:
  # Attributes and methods for Parent

  class Child:
  # Attributes and methods for Child

  class GrandChild(Parent, Child):
  # GrandChild inherits from both Parent and Child
  # GrandChild can combine or override behavior from each
  ```


## Polymorphism (Code Reuse)

- Allows methods in different classes to share names but do different tasks
  - Call same method name on different objects --> different response
- Polymorphism basic syntax:

  ```python
  class A:
      def action(self): ...

  class B:
      def action(self): ...

  class C:
      def action(self): ...

  Class().method()  # Works for A, B, or C
  ```


### Inheritance-based polymorphism - parent class defines method, multiple child classes override in own way

- Call same method on any child object & behaves differently depending on which child class
- Example:

  ```python
  class Wicket:
      def dismissal(self):
          return 'Generic wicket'

  class Pope(Wicket):
      def dismissal(self):
          return 'Falls over ball and is bowled'

  class Kohli(Wicket):
      def dismissal(self)
      return 'Nicks off to 4th stump'

  class Watson(Wicket):
      def dismissal (self)
      return 'LBW'

  print(Pope().dismissal()) # Falls over ball and is bowled
  print(Kohli().dismissal()) # Nicks off to 4th stump
  print(Watson().dismissal()) # LBW
  ```

- Could also do calling in a list & loop

  ```python
  batters = [Pope(), Kohli(), Watson()]

  for batter in batters:
      print(batter.dismissal())
  ```


## Name Mangling - prefix attribute with `__`

- `_`: convention to show private
- `__`: triggers Python's name mangling process
  - Python renames the attribute (`__attribute` --> `_ClassName__attribute`)
  - Prevents accidental attribute & method overriding during inheritance:
    - If attribute only meant for internal use in class: Use `_`
    - If class will be inherited: Use `__` to stop overriding


### Example part 1: Using \_\_ to prevent overriding


````python
class Parent:
    def __init__(self):
        self.data = "Parent Data"

class Child(Parent):
    def __init__(self):
        super().__init__()
        self.data = "Child Data" # Parent overwritten

c = Child()
print(c.__dict__) 
````

````python
class Parent:
    def __init__(self):
        self.__data = "Parent Data"

class Child(Parent):
    def __init__(self):
        super().__init__()
        self.__data = "Child Data" # __ prevents overwriting

c = Child()
print(c.__dict__)
````
