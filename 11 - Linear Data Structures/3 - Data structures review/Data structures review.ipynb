# Data Structures Review


# When to Use Each Data Structure

- Lists: When you need ordered, indexed access and don't know size in advance
- Stacks: For LIFO operations (undo functionality, expression evaluation, backtracking)
- Queues: For FIFO operations (task scheduling, breadth-first search)
- Linked Lists: When frequent insertion/deletion at beginning, unknown size, no random access needed
- Hash Maps: For fast key-value lookups, counting occurrences, caching
- Sets: For uniqueness checking, mathematical set operations, removing duplicates


## Algorithms & Big O Notation

- Algorithms: set of unambiguous instructions for solving a problem
  - Finite number of steps, each precise & unambiguous
- Big O Notation: Described worst-case performance of algorithm as input size inc
  - Resource usage (time & space) of algorithm as input size inc
  - Ignores constant factors & lower order terms `(e.g., O(n^2 + n) = O(n^2))`


### Common Time Complexities

- `O(1)`: Constant time, same time regardless of input size
- `O(log n)`: Logarithmic time, time inc slowly as input size inc.
  - Example: Binary search
- `O(n)`: Linear time, time inc linearly with input size
- `O(n log n)`: Log-linear time
  - Example: Efficient sorting algorithms (e.g., mergesort, heapsort)
- `O(n^2)`: Quadratic time, time inc quadratically with input size
  - Example: Nested Loops


### Space Complexity

- `O(1)`: Constant Space, same memory regardless of input size.
- `O(n)`: Linear Space, Memory grows proportionally with input size.
- `O(n²)`: Quadratic Space, Memory usage grows quadratically with input size.


## Problem Solving Techniques


- Understanding the Problem: Read the problem statement multiple times. Identify input, expected output, and how to transform input to output.

- Pseudocode: High-level description of algorithm logic that is language-independent. Uses common written language mixed with programming constructs like IF, ELSE, FOR, WHILE.

- Edge cases: Specific, valid inputs that occur @ boundary of what algo should handle.
  - Always consider & test


## Arrays


- Static: Fixed size @ initialisation - Elements stored in adjacent memory locations - Size can't be changed during program execution
  <br></br>
- Dynamic: Can grow/shrink automatically during program execution
  - Handle resizing by auto copyting to larger array when full


### Python lists - dynamic arrays


```python
numbers = [3, 4, 5, 6]

# Access elements
numbers[0]  # 3

# Update elements
numbers[2] = 16

# Add elements
numbers.append(7)
numbers.insert(3, 15)  # Insert at specific index

# Remove elements
numbers.pop(2)  # Remove at specific index
numbers.pop()   # Remove last element
```

### Time complexities for dynamic arrays


- Access: `O(1)`
- Insert at end: `O(1)` average, `O(n)` when resizing needed
- Insert in middle: `O(n)`
- Delete: `O(n)` for middle, `O(1)` for end


## Stacks


### LIFO data structure

- Last In First Out
- Elements added and removed from the top of the stack
- Push: Add element to top of stack
- Pop: Remove element from top of stack


```python
# Using Python list as stack
stack = []

# Push operations
stack.append(1)
stack.append(2)
stack.append(3)

# Pop operations
top_element = stack.pop()  # Returns 3
```

## Queues


### FIFO data structure

- First In First Out
- Elements addred to back and removed from front
- Enqueue: Add element to back of queue. `O(1)`
- Dequeue: Remove element from front of queue. `O(1)`


```python
from collections import deque

# Using deque for efficient queue operations
queue = deque()

# Enqueue operations
queue.append(1)
queue.append(2)
queue.append(3)

# Dequeue operations
first_element = queue.popleft()  # Returns 1
```

## Linked Lists


### Linear data structure

- Each node has data & ref to next node, linked like chain
  <br></br>
- Singly linked list - Struct: Each node has data & 1 ref to next node - Traveral: Only forward head to tail - Head node: First node in list, usually only directly accessible node - Tail node: Last node in list, points to `None`
  <br></br>
- Operations & time complexities - Insert at beginning: `O(1)` - Insert at end: `O(n)` - must traverse to end - Insert in middle: `O(n)` - must traverse to position - Delete from beginning: `O(1)` - Delete from end: `O(n)` - must traverse to find previous node - Delete from middle: `O(n)` - must traverse to find node
  <br></br>
- Doubly linked list:
  - Structure: Each node has data and two references: next node and previous node.
  - Traversal: Can move in both directions.
  - Memory: Requires more memory than singly linked lists due to extra reference.


## Hash Maps & Sets


### Hash Maps

- Map (Abstract Data Type): Manages key-value pairs
  - Unique keys
  - Non-unique values
- Hash Map: Concrete implementation of map ADT using hashing
  - Hash function: Converts key to index in underlying array
  - Determines storage location for KV pairs


#### Python Dictionaries - hash maps


```python
# Creating dictionaries
my_dictionary = {
    "A": 1,
    "B": 2, 
    "C": 3
}

# Alternative creation
my_dictionary = dict(A=1, B=2, C=3)

# Access and modify
value = my_dictionary["A"]  # 1
my_dictionary["A"] = 4      # Update value
del my_dictionary["A"]      # Remove key-value pair

# Check membership
"C" in my_dictionary

# Get keys, values, items
my_dictionary.keys()
my_dictionary.values()
my_dictionary.items()
```

#### Hash Maps time complexities

- Average case: `O(1)` for insert, get, delete
- Worst case: `O(n)` when many hash collisions occur (same index for multiple keys)


### Sets

- Unordered collection of unique elements
- Immutable elements only (numbers, strings, tuples): Hash values must be constant


```python
# Creating sets
numbers = {1, 2, 3, 4}
empty_set = set()  # Must use set(), not {}

# Add and remove elements
numbers.add(5)
numbers.remove(4)      # Raises KeyError if not found
numbers.discard(4)     # No error if not found

# Set operations
set_a = {1, 2, 3, 4}
set_b = {2, 3, 4, 5, 6}

# Union, intersection, difference, symmetric difference
set_a.union(set_b)                    # or set_a | set_b
set_a.intersection(set_b)             # or set_a & set_b
set_a.difference(set_b)               # or set_a - set_b
set_a.symmetric_difference(set_b)     # or set_a ^ set_b

# Subset and superset checks
set_a.issubset(set_b)
set_a.issuperset(set_b)
set_a.isdisjoint(set_b)

# Membership testing
5 in numbers
```

#### Time complexities for sets

- Average case: `O(1)` for insert, delete, membership test
- Worst case: `O(n)` (hash collisions)


### Hash Collisions

- 2 different keys produce same has value
- Collision resolution strategies:
  - Chaining:: Each array index points to a linked list storing all elements with same hash value
  - Open Addressing: Search for next available index using predefined sequence
