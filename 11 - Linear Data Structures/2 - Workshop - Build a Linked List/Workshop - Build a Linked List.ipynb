# Workshop - Build a Linked List

Linked list: List with nodes that contain data & ref to next node


## Construction


```python
class LinkedList:
    class Node: # Node in the LL. Element parameter is data in node
        def __init__(self, element):
            self.element = element
            self.next = None # Pointer to next node in LL. None here means no next node (end of list)
            
    def __init__(self): # Init for LL. Just self because no other parameters when calling 
        self.length = 0 # Length 0 bc no nodes yet
        self.head = None # None bc no nodes yet

    def is_empty(self):
        return self.length == 0 # Defaults to True if length is 0, False if not
    
    def add(self, element): # Add a node to the end of the list. Element is data in node
        node = self.Node(element) # Create a new node with the given element. `node` is left of `=` bc its variable to hold the new node
        if self.is_empty():
            self.head = node # sets the node we're adding to head if empty
        else:
            current_node = self.head # Start at the head of the list
            while current_node.next is not None: # While loop: runs until we reach the end of the list (when current_node.next is None)
                current_node = current_node.next # Move to next node
            current_node.next = node # Sets the pointer of the next node to the new node we're adding
        self.length += 1

    def remove(self, element): # Remove a node from the list. Element is data in node
        previous_node = None # Makes a variable to hold the previous node. Starts as None bc we haven't traversed the list yet
        current_node = self.head # Start at the head of the list
        while current_node is not None and current_node.element != element: # While loop: runs until we reach the end of the list (when current_node is None) or we find the node with the given element
            previous_node = current_node # Move one along
            current_node = current_node.next # Move one along
        if current_node is None: # If we reached the end of the list without finding the element, do nothing
            return        
        elif previous_node is not None:
            previous_node.next = current_node.next # If we found the target node, cut it out the pointer chain
        else:
            self.head = current_node.next # If the target node is the head, set the head to the next node
        self.length -= 1
```

## Using the list


```python
my_list = LinkedList()
print(my_list.is_empty())

my_list.add(1)
my_list.add(2)
print(my_list.is_empty())
print(my_list.length)

my_list.remove(1)
print(my_list.length)
```
