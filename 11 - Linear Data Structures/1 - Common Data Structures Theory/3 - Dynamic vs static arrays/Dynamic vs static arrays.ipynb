# Dynamic vs static arrays


## Static arrays

- Fixed size
- Size determines when initialised, cannot be changed
- Store elements in adjacent memory locations
  - Makes retrieval more efficient:
    - Program can store location of first element then use indices for others
    - Takes constant time `O(1)`


## Dynamic arrays (Python)

- More flexible - can grow/shrink while running
- Automatic resizing mechanism
- Accessing elemennts still `O(1)`
- Inserting element in middle `O(n)` because elements after need to be shifted
- Inserting at end is `O(1)` if space in array, but can be `O(n)` if array needs to be resized

- `list` in Python is a dynamic array (`numbers = [3, 4, 5, 6]`)
  - Update values: `numbers [2] = 16`
  - Add values: `numbers.append(7)`
  - Insert values: `numbers.insert(2, 10)` (inserts 10 at index 2)
  - Remove values: `numbers.pop(2)` (removes value at index 2)
