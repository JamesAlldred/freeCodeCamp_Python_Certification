# Singly vs double linked lists

- Linked list: Linear data structure where each node is connected to the next node in the sequence
- Chain of noodes, each containing data & a reference to the next node
- Implement other data structures


## Singly linked list

- Linked list where each node connected to next in seq
  - Connected by storing a reference
    - Allows to traverse linked list in one direction (start --> end)
    - Tail node: Last node. Determines when process has reached end of linked list


## Inserting nodes

- Linked lists don't have fixed size: can update connections
- To insert node at start:
  - Create new node
  - Set new node's next reference to current head node
  - Time complexity `O(1)`
- To insert node at end:
  - Create new node
  - Traverse linked list to find tail node: Time complexity `O(n)`
  - Set tail node's next reference to new node
- To insert node in middle:
  - Create new node
  - Traverse linked list to find node before insertion point
  - Set new node's next reference to current node's next reference
  - Set current node's next reference to new node
  - Time complexity `O(1)`: only requires creation & updating connections. Doesn't depend on size of linked list

## Removing nodes

- To remove node at start:
  - Set head reference to current head node's next reference
  - Time complexity `O(1)`
- To remove from middle:
  - Update previous node's next reference to current node's next reference
  - Time complexity `O(1)`: only requires updating connections. Doesn't depend on size of linked list
- To remove from end:
  - Traverse linked list to find node before tail node: Time complexity `O(n)`
  - Remove previous node's reference
  - The deletion operation has constant space complexity `O(1)`
    - No additional memory required to delete a node


## Doubly linked list

- Each node stores 2 references: next & previous
  - Traversal both directions
- Common to keep rederence to tail node in linked list itself: start traversal from end if needed
- More memory needed (2 references per node)
