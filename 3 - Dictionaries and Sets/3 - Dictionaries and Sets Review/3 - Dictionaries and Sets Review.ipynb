```python
###Dictionaries - built in data structures that store collections of key-value pairs
#Keys need to be immutale
dictionary = {
    'key1': 'value1',
    'key2': 'value2'



#dict() - build dictionary. Pass list of tuples as argument. key, value
pizza = dict([('name', 'Margherita Pizza'), ('price', 8.9), ('calories_per_slice', 250), ('toppings', ['mozzarella', 'basil'])])
print(pizza)
```

```python
###Common dictionary methods

#.get(key, default) - returns value associated with key. can set defaut - prevents error if key doesnt exist
#.keys() - views all keys
#.values() - views all values
#.items() - views all key-value pairs
#.clear() - removes all key-value pairs from dictionary
#.pop() - removes last inserted item
#.update() - update with key-value pairs with k-v pairs of other dict. Overwritten if in common, added if new
```

```python
###Looping over a dictionary
products = {
    'Laptop': 990,
    'Smartphone': 600,
    'Tablet': 250,
    'Headphones': 70,
}
#iterating over values - for loop with .values()
for price in products.values():
    print(price)

#iterating over keys - for loop with .keys() OR [dictionary name] directly
for product in products:
    print(product)
    
#iterating over key-value pairs - for loop with .items(). Gets tuples w/ keys + values
for pair in products.items():
    print(pair)
#to separate key & value - comma. 
for a,b in products.items():
    print(a,b)
#enumerate for index. pass 2nd argument to start from custom value
for x, y in enumerate(products.items(),  1):
    print(x, y)
```

```python
###sets - built in data structures. No duplicates, mutable, unordered.
#only store immutable data types
#define with curly brackets or set() if empty
my_set = {1, 2, 3, 4}
set()
```

```python
###common set methods

#.add() - adds element to set
#.remove - removes element. KeyError if not found
#.discard() - removes element. no error
#.clear() - removes all elements
```

```python
###mathematical set operations

#.issubset()
#.issuperset
my_set = {1, 2, 3, 4, 5}
your_set = {2, 3, 4, 5}

print(your_set.issubset(my_set))
print(my_set.issuperset(your_set))

#.isdisjoint() - check if they don't have common elements
set1 = {1,2}
set2 = {3,4}

print(set1.isdisjoint(set2))

#Union operator (|) - returns new set with elements from both
set3 = {5,6}
set4 = {7,8}

print(set3 | set4)

#Intersection operator (&) - returns new set of commmon elements
set5 = {9,10}
set6 = {10,11}

print(set5 & set6)

#difference operator (-) - returns new set of first set that arent in other
set7 = {12,13}
set8 = {13,14}
print(set7 - set8)

#Symmetric Difference Operator (^) - returns set of elements from either set but not both
set9 = {1,2}
set10 = {1,3}
print(set9^set10)

#in operator - check if element in set
set11 = {1,2}
print(1 in set11)
```

__name__ Variable: __name__ is a special built-in variable in Python. When a Python file is executed directly, Python sets the value of this variable to the string "__main__". But if the Python file is imported as a module into another Python script, the value of the __name__ variable is set to the name of that module.

This is why you'll often find this conditional in Python scripts. It contains the code that you only want to run only if the Python script is running as the main program.
